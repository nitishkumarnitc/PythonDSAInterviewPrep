# Technical Problem Solving – Model Answers (Part B)

Use this after you have attempted all problems in `dsa_prep.ipynb`. For each
one, we sketch the core approach plus time and space complexity. In an
interview, you should be able to derive this live and justify correctness.

## Section 1 – Arrays and basic algorithms

1. Two Sum – **Approach**: Use a hash map from value to index. As you scan the
   array once, for each `x` check if `target - x` is already in the map; if
   yes, return that index and current index, otherwise insert `x`. **Time**:
   O(n). **Space**: O(n) for the hash map.
2. Maximum Subarray – **Approach**: Kadane's algorithm: maintain
   `current_max = max(nums[i], current_max + nums[i])` and
   `global_max = max(global_max, current_max)` while scanning once. **Time**:
   O(n). **Space**: O(1).
3. Product of Array Except Self – **Approach**: Precompute prefix products
   (left to right) and suffix products (right to left);
   `answer[i] = prefix[i-1] * suffix[i+1]`. Alternatively, do it in-place with
   one prefix pass and one suffix-running-product pass. Handle zeros by
   counting how many zeros there are. **Time**: O(n). **Space**: O(1) extra
   (beyond output) if done carefully.
4. Merge Sorted Arrays – **Approach**: If we have extra space at the end of
   the first array, fill from the back: maintain three pointers at the ends of
   both arrays and the write index, and always place the larger of the two
   current elements. **Time**: O(m + n). **Space**: O(1) extra.

## Section 2 – Strings and hashing

5. Valid Anagram – **Approach**: Either sort both strings and compare
   (O(n log n)), or more efficiently, count characters in a fixed-size
   array/dict and ensure counts match. **Time**: O(n) with counting.
   **Space**: O(1) / O(Σ) for fixed alphabet, or O(k) where k is number of
   distinct chars.
6. Group Anagrams – **Approach**: Build a hash map from a canonical key for
   each word to a list of words. Common keys: sorted characters
   (`''.join(sorted(s))`) or a 26-count tuple for letters. **Time**:
   O(n * L log L) with sorting per word, or O(n * L) with counting (n = words,
   L = avg length). **Space**: O(n * L) for storing groups and keys.
7. Longest Substring Without Repeating Characters – **Approach**: Sliding
   window with a hash map `last_index` of characters. Expand the right
   pointer; if you see a repeated char inside the window, move the left
   pointer to `max(left, last_index[char] + 1)`. Track max window size.
   **Time**: O(n). **Space**: O(min(n, Σ)) for the map.

## Section 3 – Two pointers and sliding window

8. Container With Most Water – **Approach**: Two pointers at both ends;
   compute area = `min(height[l], height[r]) * (r - l)`, update best. Move the
   pointer at the shorter line inward (that’s the only way to possibly
   increase area). **Time**: O(n). **Space**: O(1).
9. Minimum Window Substring – **Approach**: Sliding window with frequency
   maps. Maintain required counts from `t` and current counts in the window;
   expand right pointer and once all requirements are met, try to shrink from
   the left while still valid, tracking the minimum window. **Time**:
   O(|s| + |t|). **Space**: O(Σ) / O(k) for counts.
10. Smallest Subarray With Sum ≥ S – **Approach**: Because all numbers are
    positive, use a sliding window: expand right and add to a running sum;
    when sum ≥ S, move left forward subtracting from sum and update minimal
    length. **Time**: O(n). **Space**: O(1).

## Section 4 – Linked lists and stacks

11. Linked List Cycle II – **Approach**: Use Floyd's Tortoise and Hare to
    detect a cycle (fast and slow pointers). When they meet, put one pointer
    at head, then move both one step at a time; the node at which they meet
    again is the cycle start. **Time**: O(n). **Space**: O(1).
12. Merge K Sorted Lists – **Approach**: Use a min-heap (priority queue)
    initialized with the head of each non-empty list; repeatedly pop the
    smallest node, append to the result, and push its next node. **Time**:
    O(N log k), where N is total nodes, k is number of lists. **Space**:
    O(k) for the heap.
13. Min Stack – **Approach**: Maintain two stacks: one for values, one for
    current minimums; on push, also push `min(x, current_min)` to the
    min-stack; on pop, pop both. Top of min-stack is O(1) minimum. **Time**:
    O(1) per operation. **Space**: O(n).

## Section 5 – Trees and graphs

14. Binary Tree Level Order Traversal – **Approach**: Standard BFS with a
    queue; process each level by taking the current queue size, popping that
    many nodes, and pushing their children. **Time**: O(n). **Space**: O(n) in
    worst case for the queue.
15. Validate Binary Search Tree – **Approach**: Either do an inorder
    traversal and verify the resulting sequence is strictly increasing, or use
    recursion with `(min_allowed, max_allowed)` bounds for each node. **Time**:
    O(n). **Space**: O(h) for recursion stack (h = height).
16. Lowest Common Ancestor – **Approach (general tree)**: Recurse from root;
    if current node is `p` or `q`, return it; otherwise recurse into
    left/right. If both sides return non-null, current node is LCA; else
    propagate non-null result up. **Approach (BST)**: Exploit ordering: if
    both `p` and `q` values are < root, go left; if both > root, go right;
    otherwise root is LCA. **Time**: O(n) for general, O(h) for BST.
    **Space**: O(h) recursion.
17. Number of Islands – **Approach**: Iterate over all cells; when you see
    a '1' not yet visited, increment island count and run DFS/BFS to mark all
    reachable '1's as visited (or flip to '0'). **Time**: O(m * n). **Space**:
    O(m * n) worst case for recursion/queue or visited set.

## Section 6 – Dynamic programming and greedy

18. Coin Change (Min Coins) – **Approach**: Bottom-up DP:
    `dp[a] = min over coins c of (dp[a - c] + 1)` with base `dp[0] = 0`,
    initialize others to INF; final answer is `dp[amount]` or -1 if
    unreachable. Top-down memoized recursion is equivalent. **Time**:
    O(amount * num_coins). **Space**: O(amount).
19. Longest Increasing Subsequence – **Approach (O(n^2))**: DP where
    `dp[i]` is LIS ending at `i`, looking back at all `j < i` with
    `nums[j] < nums[i]`. **Approach (O(n log n))**: Maintain an array `tails`
    where `tails[len]` is the smallest tail value of an increasing subsequence
    of length `len + 1`; for each `x`, binary search `tails` for where to
    place it. **Time**: O(n^2) or O(n log n). **Space**: O(n).
20. Interval Scheduling – **Approach**: Sort intervals by end time
    ascending, then greedily pick the first interval, and then always pick the
    next interval with start >= `last_end`. This greedy choice is optimal
    because picking the earliest finishing interval leaves maximal room for
    subsequent intervals. **Time**: O(n log n) for sorting. **Space**: O(1) or
    O(n) depending on how you store output.


## Detailed solutions with code examples (Problems 1-3)

Below are more detailed notes and reference implementations for the first three

### Problem 1 - Two Sum

**Idea:**

- Scan the array once, keeping a hash map `seen` from value -> index.
- For each `x` at index `i`, compute `need = target - x`.
- If `need` is already in `seen`, you found a pair; return `[seen[need], i]`.
- Otherwise store `seen[x] = i` and continue.

**Complexity:** `O(n)` time, `O(n)` extra space for the hash map.


```python
from typing import List


def two_sum(nums: List[int], target: int) -> List[int]:
    """Return indices of two numbers that add up to target.

    Time: O(n), Space: O(n).
    """

    seen = {}
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [seen[need], i]
        seen[x] = i
    return []


print(two_sum([2, 7, 11, 15], 9))
```

### Problem 2 - Maximum Subarray (Kadane)

**Idea:**

- Maintain two running values while scanning left to right:
  - `current_max` = best subarray sum **ending at** current index.
  - `global_max` = best subarray sum seen so far anywhere.
- Recurrence: `current_max = max(nums[i], current_max + nums[i])`.
- Update `global_max = max(global_max, current_max)`.

**Complexity:** `O(n)` time, `O(1)` extra space.


```python
def max_subarray(nums: List[int]) -> int:
    """Return the maximum subarray sum using Kadane's algorithm.

    Time: O(n), Space: O(1).
    """

    current_max = global_max = nums[0]
    for x in nums[1:]:
        current_max = max(x, current_max + x)
        global_max = max(global_max, current_max)
    return global_max


print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```

### Problem 3 - Product of Array Except Self

**Idea:**

- We want `answer[i]` = product of all elements except `nums[i]`.
- Do a prefix pass to build products to the left of each index.
- Then walk from the right, keeping a running suffix product, and multiply it
  into the prefix for each index.
- This avoids division and uses only `O(1)` extra space beyond the output.

**Complexity:** `O(n)` time, `O(1)` extra space (besides output array).


```python
def product_except_self(nums: List[int]) -> List[int]:
    """Return array of products except self without using division.

    Time: O(n), Space: O(1) extra (output not counted).
    """

    n = len(nums)
    res = [1] * n

    # Prefix products
    prefix = 1
    for i in range(n):
        res[i] = prefix
        prefix *= nums[i]

    # Suffix products
    suffix = 1
    for i in range(n - 1, -1, -1):
        res[i] *= suffix
        suffix *= nums[i]

    return res


print(product_except_self([1, 2, 3, 4]))
```

### Problem 4 - Merge Sorted Arrays

**Idea:**

- Assume the classic interview variant where `nums1` has length `m + n` with
  the first `m` elements valid and extra space at the end to hold `nums2`.
- Use three pointers starting from the **end**: one at `m - 1` in `nums1`, one
  at `n - 1` in `nums2`, and a write index at `m + n - 1`.
- At each step, place the larger of the two pointed values into the write
  position and move pointers left. This avoids overwriting untouched data.

**Complexity:** `O(m + n)` time, `O(1)` extra space.


```python
from typing import List


def merge_sorted_arrays(nums1: List[int], m: int, nums2: List[int], n: int) -> None:
    """Merge nums2 into nums1 in-place assuming nums1 has enough trailing space.

    Time: O(m + n), Space: O(1).
    """

    i = m - 1
    j = n - 1
    write = m + n - 1

    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[write] = nums1[i]
            i -= 1
        else:
            nums1[write] = nums2[j]
            j -= 1
        write -= 1


arr1 = [1, 3, 5, 0, 0, 0]
arr2 = [2, 4, 6]
merge_sorted_arrays(arr1, 3, arr2, 3)
print(arr1)
```

### Problem 5 - Valid Anagram

**Idea:**

- Two strings are anagrams if they contain the same multiset of characters.
- Either sort both strings and compare (O(n log n)), or count characters in a
  dictionary/array and compare counts (O(n)).

**Complexity:** `O(n)` time and `O(Σ)` space with counting, where Σ is the
alphabet size.


```python
def is_anagram(s: str, t: str) -> bool:
    """Return True if s and t are anagrams of each other.

    Time: O(n), Space: O(Σ).
    """

    if len(s) != len(t):
        return False

    counts = {}
    for ch in s:
        counts[ch] = counts.get(ch, 0) + 1
    for ch in t:
        if ch not in counts:
            return False
        counts[ch] -= 1
        if counts[ch] == 0:
            del counts[ch]
    return not counts


print(is_anagram("listen", "silent"), is_anagram("foo", "bar"))
```

### Problem 6 - Group Anagrams

**Idea:**

- For each word, build a canonical key that is the same for all anagrams.
- Common keys: sorted tuple of characters or a 26-length frequency tuple for
  lowercase letters.
- Use a dictionary mapping key -> list of words, then return the grouped
  values.

**Complexity:** `O(n * L log L)` time with sorting (L = word length), or
`O(n * L)` with counting; `O(n * L)` space for storing the groups.


```python
from collections import defaultdict


def group_anagrams(strs: List[str]) -> List[List[str]]:
    """Group anagrams together using sorted-character keys.

    Time: O(n * L log L), Space: O(n * L).
    """

    groups: dict[str, List[str]] = defaultdict(list)
    for word in strs:
        key = "".join(sorted(word))
        groups[key].append(word)
    return list(groups.values())


print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

### Problem 7 - Longest Substring Without Repeating Characters

**Idea:**

- Use a sliding window `[left, right]` over the string and a map from
  character -> last seen index.
- When you see a repeated character `c` at index `right`, move `left` to
  `max(left, last_index[c] + 1)` to ensure the window has no duplicates.
- Track the maximum window length as you go.

**Complexity:** `O(n)` time and `O(min(n, Σ))` space.


```python
def length_of_longest_substring_no_repeat(s: str) -> int:
    """Return length of longest substring without repeating characters.

    Time: O(n), Space: O(min(n, Σ)).
    """

    last_index: dict[str, int] = {}
    left = 0
    best = 0

    for right, ch in enumerate(s):
        if ch in last_index and last_index[ch] >= left:
            left = last_index[ch] + 1
        last_index[ch] = right
        best = max(best, right - left + 1)
    return best


print(length_of_longest_substring_no_repeat("abcabcbb"))
```

### Problem 8 - Container With Most Water

**Idea:**

- Place two pointers at the ends of the array representing line heights.
- The area is limited by the shorter line, so move the pointer at the shorter
  line inward, hoping to find a taller line that might increase area.
- Continue until pointers meet, tracking the maximum area.

**Complexity:** `O(n)` time, `O(1)` extra space.


```python
def max_area(height: List[int]) -> int:
    """Return the maximum water container area using two pointers.

    Time: O(n), Space: O(1).
    """

    left, right = 0, len(height) - 1
    best = 0

    while left < right:
        h = min(height[left], height[right])
        best = max(best, h * (right - left))
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    return best


print(max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]))
```

### Problem 9 - Minimum Window Substring

**Idea:**

- Maintain counts of required characters from `t` and counts of the current
  window in `s`.
- Expand the right pointer adding characters; once the window satisfies all
  required counts, try to shrink from the left while keeping it valid,
  updating the best (shortest) window.

**Complexity:** `O(|s| + |t|)` time, `O(Σ)` space for the count maps.


```python
def min_window_substring(s: str, t: str) -> str:
    """Return the minimum window in s that contains all chars of t.

    Time: O(|s| + |t|), Space: O(Σ).
    """

    if not t or not s:
        return ""

    need: dict[str, int] = {}
    for ch in t:
        need[ch] = need.get(ch, 0) + 1

    have: dict[str, int] = {}
    required = len(need)
    formed = 0
    left = 0
    best_len = float("inf")
    best_range = (0, 0)

    for right, ch in enumerate(s):
        have[ch] = have.get(ch, 0) + 1
        if ch in need and have[ch] == need[ch]:
            formed += 1

        while formed == required:
            if right - left + 1 < best_len:
                best_len = right - left + 1
                best_range = (left, right)

            left_ch = s[left]
            have[left_ch] -= 1
            if left_ch in need and have[left_ch] < need[left_ch]:
                formed -= 1
            left += 1

    if best_len == float("inf"):
        return ""
    start, end = best_range
    return s[start : end + 1]


print(min_window_substring("ADOBECODEBANC", "ABC"))
```

### Problem 10 - Smallest Subarray With Sum ≥ S

**Idea:**

- Because all numbers are positive, the window sum increases as you move the
  right pointer right and decreases as you move the left pointer right.
- Maintain a running sum and a sliding window; whenever the sum is ≥ S, move
  the left pointer right while updating the best (smallest) window length.

**Complexity:** `O(n)` time, `O(1)` extra space.


```python
def min_subarray_len(target: int, nums: List[int]) -> int:
    """Return the minimal length of a subarray with sum at least target.

    Time: O(n), Space: O(1).
    """

    left = 0
    total = 0
    best = float("inf")

    for right, x in enumerate(nums):
        total += x
        while total >= target:
            best = min(best, right - left + 1)
            total -= nums[left]
            left += 1

    return 0 if best == float("inf") else best


print(min_subarray_len(7, [2, 3, 1, 2, 4, 3]))
```
