# Technical Problem Solving – Part B

This notebook contains 20 data-structures-and-algorithms problems in an
interview style. For each problem, write code in a new code cell and be ready
to discuss time and space complexity, edge cases, and possible optimizations.


## Section 1 – Arrays and basic algorithms


### Problem 1 – Two Sum

Given an array of integers and a target value, return indices of two numbers
that add up to the target. Discuss a hash-map-based O(n) solution and
trade-offs versus the naive O(n^2) approach.


### Problem 2 – Maximum Subarray

Given an integer array (which may contain negative numbers), find a contiguous
subarray with the maximum sum and return that sum. Explain Kadane's algorithm
and why it is O(n).


### Problem 3 – Product of Array Except Self

Given an integer array `nums`, return an array `answer` where `answer[i]` is
the product of all elements except `nums[i]`, without using division and in
O(n) time. Discuss how to handle zeros.


### Problem 4 – Merge Sorted Arrays

You are given two sorted arrays (or one large array with extra space at the
end). Merge them into one sorted array in-place. Discuss time/space complexity
and edge cases.


## Section 2 – Strings and hashing


### Problem 5 – Valid Anagram

Given two strings, determine if one is an anagram of the other. Discuss
approaches using sorting vs counting and their complexities.


### Problem 6 – Group Anagrams

Given a list of strings, group them into lists of anagrams. Explain different
ways to construct hashable keys and their trade-offs.


### Problem 7 – Longest Substring Without Repeating Characters

Given a string `s`, find the length of the longest substring without repeating
characters. Describe a sliding-window + hash-map solution and its complexity.


## Section 3 – Two pointers and sliding window


### Problem 8 – Container With Most Water

Given an array of non-negative integers representing vertical lines, choose two
lines that together with the x-axis form a container that holds the most
water. Explain the two-pointer approach and why it is correct.


### Problem 9 – Minimum Window Substring

Given strings `s` and `t`, find the smallest window in `s` that contains all
characters of `t` (with multiplicity). Describe a sliding-window solution and
discuss complexity and corner cases.


### Problem 10 – Smallest Subarray With Sum ≥ S

Given an array of positive integers and a target sum `S`, find the minimal
length of a contiguous subarray of which the sum ≥ `S`. Explain why a
sliding-window approach works here.


## Section 4 – Linked lists and stacks


### Problem 11 – Linked List Cycle II

Given the head of a singly linked list, determine if there is a cycle and, if
so, return the node where the cycle begins. Discuss Floyd's Tortoise and Hare
algorithm and the proof idea.


### Problem 12 – Merge K Sorted Lists

Given `k` sorted linked lists, merge them into one sorted list. Compare the
complexity of a naive pairwise merge vs using a min-heap / priority queue.


### Problem 13 – Min Stack

Design a stack that supports push, pop, top, and retrieving the minimum
element in constant time. Explain your internal representation and how you
handle duplicates.


## Section 5 – Trees and graphs


### Problem 14 – Binary Tree Level Order Traversal

Given the root of a binary tree, return its level order traversal (values by
level). Implement BFS and discuss its time and space complexity.


### Problem 15 – Validate Binary Search Tree

Given the root of a binary tree, determine if it is a valid BST. Compare an
inorder traversal approach vs passing valid value ranges down the recursion.


### Problem 16 – Lowest Common Ancestor

Given a binary tree (or BST) and two nodes, find their lowest common
ancestor. Explain how the solution differs between a general binary tree and a
BST.


### Problem 17 – Number of Islands

Given a 2D grid of '1's (land) and '0's (water), count the number of islands.
Outline DFS/BFS approaches and how you mark visited cells.


## Section 6 – Dynamic programming and greedy


### Problem 18 – Coin Change (Min Coins)

Given coin denominations and a total amount, compute the minimum number of
coins needed to make that amount (or report that it is impossible). Explain
both top-down (memoized) and bottom-up DP formulations.


### Problem 19 – Longest Increasing Subsequence

Given an array of integers, find the length of the longest strictly increasing
subsequence. Compare the O(n^2) DP with the O(n log n) patience-sorting-style
solution.


### Problem 20 – Interval Scheduling

Given a set of intervals with start and end times, select the maximum number
of non-overlapping intervals. Describe a greedy strategy (such as sorting by
end time) and prove why it is optimal.
