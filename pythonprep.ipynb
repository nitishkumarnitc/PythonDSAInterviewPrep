# Python Fundamentals – Part A

This notebook contains 50 targeted Python fundamentals questions for interview prep focused on top-tier engineering and quant-style roles.

## Section 1 – Core semantics and control flow
1. Names and objects: Explain how Python binds names to objects and what happens when two variables reference the same list and you mutate it.
2. Mutability and arguments: How does Python pass arguments to functions, and how do mutable vs immutable types affect visible side effects?
3. Loops vs list comprehensions vs generator expressions: Compare them in terms of readability, evaluation (eager vs lazy), and memory usage.
4. `is` vs `==`: What is the difference, and why is `x is None` preferred to `x == None`?
5. Truthiness: How does Python decide if an object is truthy or falsy? List common falsy values and mention `__bool__` / `__len__`.

## Section 2 – Built-in data structures and complexity
6. list / tuple / set / dict: Compare mutability, ordering, and average time complexity for lookup/insert/delete; when would you pick each?
7. Hashing and dict/set: What does it mean for an object to be hashable, and why must `__eq__` and `__hash__` be consistent?
8. Shallow vs deep copy: Explain with an example where shallow copy of a nested structure leads to surprising mutations.
9. collections utilities: When would you use `deque`, `defaultdict`, `Counter`, or `dataclass` instead of plain lists/dicts?
10. Iterators and generators: What is the iterator protocol and when do generators give big memory or performance benefits?

## Section 3 – OOP, modules, and interfaces
11. Attribute lookup: Describe how Python finds attributes (instance → class → base classes) and why mutable class attributes can be a bug.
12. Methods: Contrast instance methods, `@classmethod`, and `@staticmethod` in terms of signatures and when to use each.
13. Dunder methods: What roles do `__repr__`, `__str__`, `__eq__`, `__lt__`, `__len__`, and `__iter__` play in making user-defined classes feel like built-ins?
14. Modules and packages: Difference between a module and a package, and high-level summary of how `import` finds and loads modules.
15. Context managers: Conceptually, what problem do they solve? Explain what happens in `with open(path) as f:` and how `__enter__` / `__exit__` work.

## Section 4 – Exceptions, memory management, and concurrency
16. Exception handling: Explain `try/except/else/finally` and when exceptions are preferable to return-code style error handling.
17. Memory management: How does CPython use reference counting and a cyclic garbage collector; what does `del x` actually do?
18. GIL and concurrency: What is the Global Interpreter Lock and how does it affect CPU-bound vs I/O-bound multi-threaded programs?
19. Deadlocks and race conditions: Describe a simple deadlock scenario with two locks and how to avoid such issues.

## Section 5 – Debugging, profiling, and best practices
20. Debugging and profiling: Which tools/approaches would you use to debug a tricky bug and to find performance bottlenecks in Python code? Mention logging, debuggers, `timeit`, and `cProfile`.

## Section 6 – Functions, parameters, and decorators
21. Mutable default arguments: Show a function definition that has the classic mutable-default bug, explain why it happens, and give two idiomatic ways to fix it.
22. Function signatures: Explain how positional-only, keyword-only, `*args`, and `**kwargs` work, and how you would design a safe, self-documenting API using these features.
23. Closures and late binding: In a loop that builds a list of lambdas, why do all lambdas sometimes see the last loop value? Show how to fix this using default arguments or `functools.partial`.
24. Decorators: Conceptually, how do decorators work in Python? Describe or sketch a timing/logging decorator and explain why `functools.wraps` is important.
25. Caching: When would you use `functools.lru_cache` for performance? What are the trade-offs, and what pitfalls arise if arguments are mutable or not hashable?
26. Partial application: Compare using `functools.partial`, lambdas, and closures for specializing functions. When is `partial` preferable?
27. Contextlib: Show how you would implement a simple context manager using `contextlib.contextmanager` and compare it to a class-based context manager.

## Section 7 – Advanced data model and custom types
28. Properties: When would you expose something as a plain attribute vs a `@property` vs an explicit `get_*/set_*` method? What pitfalls occur if a property does heavy work?
29. Descriptors: At a high level, what is the descriptor protocol (`__get__`, `__set__`, `__set_name__`)? Give an example use case such as computed attributes or validation.
30. `__slots__`: How do `__slots__` change object layout and memory usage? What trade-offs and limitations come with using them (e.g., inheritance, lack of `__dict__`)?
31. Dynamic attribute access: Contrast `__getattr__` and `__getattribute__`. When would you use them (for example, proxies or lazy loading), and how do you avoid infinite recursion?
32. Custom containers: If you implement your own container type, which dunder methods do you need for it to behave like a sequence (`len`, indexing, iteration, membership)?
33. Dataclasses vs `namedtuple` vs plain classes: Compare them in terms of mutability, ergonomics, performance, and typical use cases.
34. Rich comparisons: How do `__eq__`, `__lt__`, and the other rich comparison methods work together, and what does `functools.total_ordering` do for you?
35. Hash and equality for custom types: How do you design `__hash__` and `__eq__` for a class intended to be used as a dict key or in a set, and why is mutability dangerous here?

## Section 8 – Iterators, generators, and streaming
36. Iterator exhaustion: What happens when you iterate over an iterator twice? How would you design an API that accepts “iterables” without accidentally consuming them multiple times?
37. Generator cleanup: How can a generator ensure resources are released even if the caller stops consuming early? Explain the role of `try/finally`, `.close()`, and garbage collection.
38. `yield from`: What does `yield from` do, and when is it preferable to a manual `for` loop yielding from a sub-generator?
39. Streaming pipelines: How would you process a very large file or stream (many GB) in Python so that memory usage stays bounded? Mention iterators, generators, `itertools`, and context managers.
40. Batching: How would you design an iterator or generator that yields items in fixed-size batches for downstream consumers (for example, database writes or network calls)?

## Section 9 – Concurrency and async
41. Concurrency models: Compare threads, processes, and async IO in Python. Under the GIL, when is each model appropriate (CPU-bound vs I/O-bound vs high-concurrency network servers)?
42. `concurrent.futures`: How would you use `ThreadPoolExecutor` or `ProcessPoolExecutor` to parallelize work? Explain how you’d handle exceptions and partial failures.
43. Asyncio: At a high level, how does `asyncio` work (event loop, coroutines, tasks)? Give an example of when you’d choose `asyncio` over threads.
44. Shared state and locks: Why doesn’t the GIL alone make all multi-threaded code safe? Explain how you’d protect shared mutable state and avoid deadlocks.
45. Rate limiting / throttling: How might you implement a rate-limited client (for example, HTTP requests per second) in Python using threads or `asyncio` primitives?

## Section 10 – Tooling, testing, and performance
46. Logging: How does the `logging` module’s logger hierarchy work, and how do you structure logging in a library vs in an application? Why is logging preferable to `print` in production code?
47. Testing: How would you structure Python code to be testable (for example, dependency injection and avoiding global state), and what tools/features (`unittest`, `pytest`, `unittest.mock`) would you rely on?
48. Type hints: What are the benefits and limitations of type hints in Python? How would you introduce static type checking (such as `mypy`) into an existing codebase?
49. Performance analysis: Given a slow Python service, how would you identify whether the bottleneck is algorithmic vs Python overhead? Mention `cProfile`, line profilers, and careful micro-benchmarking.
50. Beyond pure Python: When a hotspot remains even after algorithmic fixes, what options do you have (for example, optimized libraries like NumPy, C extensions, Cython, Numba, or moving work to another service), and what trade-offs do they involve?
