# Python Fundamentals – Model Answers (Part A)

Use this after you have attempted all questions in `pythonprep.ipynb`. Answers are intentionally compact – expand verbally in practice.

1. Names and objects: Names are references to objects; assignment rebinds names, not copies. If `a` and `b` both reference the same list and you mutate it (for example, `a.append(4)`), both see the change.
2. Mutability and arguments: Python is call-by-object-reference. Rebinding a parameter doesn’t affect the caller, but mutating a passed-in mutable object does. Immutables (int/str/tuple) can’t be changed in place.
3. Loops vs comprehensions vs generators: For-loops are explicit and flexible; list comprehensions are concise but allocate a whole list; generator expressions are lazy and memory efficient for large or streaming data.
4. `is` vs `==`: `==` tests value equality via `__eq__`; `is` tests identity. Use `is None` to check for the singleton `None`.
5. Truthiness: `bool(x)` uses `__bool__` or `__len__`. Falsy: `False`, `0`, `0.0`, `""`, `[]`, `{}`, `set()`, `None`. Custom classes can define `__bool__`/`__len__`.
6. list/tuple/set/dict: list – mutable, ordered, O(1) append, O(1) index; tuple – immutable list; set – unique hashable elements, average O(1) membership; dict – key->value map, average O(1) lookup/insert/delete.
7. Hashing: Hashable objects have stable `__hash__` and compatible `__eq__` (if `a == b` then `hash(a) == hash(b)`). Dicts/sets are hash tables; inconsistency breaks lookups.
8. Copying: Shallow copy duplicates the outer container but shares inner objects; deep copy recursively copies nested objects. Shallow copy is dangerous when you mutate nested structures.
9. collections utilities: `deque` – fast appends/pops from both ends (queues/BFS); `defaultdict` – auto-initialize missing keys; `Counter` – counting; `dataclass` / `namedtuple` – lightweight records with named fields.
10. Iterators/generators: Iterator protocol = `__iter__` + `__next__`. Generators yield values lazily, reducing memory for large sequences or streams.
11. Attribute lookup: Instance dict -> class -> base classes (MRO). Mutable class attributes are shared across instances and often a bug when used as default containers.
12. Methods: Instance methods get `self` (instance), `@classmethod` gets `cls` (class, good for alternative constructors), `@staticmethod` gets neither (utility function namespaced in the class).
13. Dunder methods: `__repr__` (unambiguous, debug), `__str__` (user-friendly), comparison dunders, `__len__`, `__iter__` integrate with `print`, `len`, sorting, iteration.
14. Modules/packages: Module = single `.py` file; package = directory acting as a module (traditionally with `__init__.py`). Import searches `sys.meta_path` / `sys.path` and caches in `sys.modules`.
15. Context managers: Provide `__enter__` and `__exit__` to manage resources (open/close, acquire/release) reliably even on exceptions (for example, files, locks, timers).
16. Exceptions: `try/except/else/finally` cleanly separates normal flow from error handling and cleanup. Exceptions are preferred over manual error codes in idiomatic Python.
17. Memory: CPython uses reference counting plus a cyclic GC. `del x` deletes a name binding; when refcount hits zero, the object can be freed, though actual OS memory reuse is allocator-dependent. Cycles require GC.
18. GIL: Only one thread runs Python bytecode at a time in CPython, so CPU-bound threads don’t scale across cores. I/O-bound threads and multiprocessing/asyncio are better for concurrency.
19. Deadlocks/races: Deadlocks arise when threads hold locks and wait on each other; avoid by consistent lock ordering, small critical sections, timeouts, and higher-level thread-safe primitives (queues).
20. Debugging/profiling: Use structured logging, `pdb`/IDE debuggers, minimal repro cases; for performance use `timeit`, `cProfile`, line profilers; write tests and follow PEP 8 and type hints for maintainable code.
21. Mutable defaults: A mutable default (like a list or dict) is created once at function definition time and reused across calls, so mutations leak between calls; fix by using `None` as the default and initializing inside the function, or by using immutable defaults/dataclasses.
22. Function signatures: Positional-only (`/`) and keyword-only (`*`) parameters let you control how arguments are passed; combined with `*args`/`**kwargs`, they make public APIs explicit, self-documenting, and stable as they evolve.
23. Closures and late binding: Loop variables are looked up when the lambda is called, not when it is defined, so all lambdas may see the final loop value; capture the value via a default argument (for example, `lambda x=i: ...`) or with `functools.partial`.
24. Decorators: A decorator is a callable that takes a function and returns a new callable; use it for cross-cutting concerns (logging, timing, auth). Apply `functools.wraps` to preserve the wrapped function’s name, docstring, and other metadata.
25. Caching: `functools.lru_cache` memoizes pure, deterministic functions with hashable arguments, trading memory for speed; it is unsafe when arguments or relevant state are mutable or when results depend on external side effects.
26. Partial application: `functools.partial` creates a new callable with some arguments pre-filled; it is often clearer and more reusable than ad-hoc lambdas or closures when you simply want to bind a few parameters.
27. Contextlib: `@contextmanager` lets you write a generator that `yield`s once and wraps it as a context manager, which is concise for simple setup/teardown compared with writing a full class.
28. Properties: Use plain attributes for simple data, `@property` when you need computed or validated access with an attribute-like API; avoid heavy or side-effectful properties that surprise callers or make performance unpredictable.
29. Descriptors: Objects implementing `__get__`/`__set__` (like `property`) that control attribute access at the class level; they are useful for reusable behaviors such as validation, lazy loading, or computed fields shared across multiple classes.
30. `__slots__`: Declaring `__slots__` removes the per-instance `__dict__`, saving memory and speeding attribute access, but it restricts which attributes can exist and complicates inheritance, multiple inheritance, and pickling.
31. Dynamic attribute access: `__getattr__` is called only for missing attributes, while `__getattribute__` intercepts *all* lookups; proxies often override `__getattribute__` but must delegate to `super().__getattribute__` to avoid infinite recursion.
32. Custom containers: To behave like a sequence, implement `__len__`, `__getitem__`, and usually `__iter__`/`__contains__`; this makes your type work with `len`, iteration, slicing, and `in` checks.
33. Dataclasses vs `namedtuple` vs plain classes: `namedtuple` is immutable, lightweight, and indexable; dataclasses can be mutable or frozen and auto-generate methods; plain classes give full control but require more boilerplate.
34. Rich comparisons: Implement `__eq__` and one ordering method such as `__lt__`; `functools.total_ordering` can synthesize the rest, but you must ensure the comparison defines a consistent total order.
35. Hash and equality: Hashable types must have `__hash__` consistent with `__eq__` and rely only on immutable state; if you mutate fields that participate in hashing, dict/set membership and lookups can silently break.
36. Iterator exhaustion: Iterators are one-shot; once consumed they raise `StopIteration` on further use, so APIs that may need multiple passes should either materialize data or accept a general iterable and document consumption carefully.
37. Generator cleanup: Use `try/finally` in generators to release resources when iteration stops early; the runtime calls `close()` (and executes `finally`) when the generator is garbage-collected or explicitly closed.
38. `yield from`: `yield from subgen` delegates iteration to a sub-generator, forwarding values, `send`, and exceptions; it simplifies generator composition and improves readability over manually looping and yielding.
39. Streaming pipelines: Process large inputs line-by-line or chunk-by-chunk using iterators/generators and `itertools`, combined with context managers so only small pieces are in memory at a time.
40. Batching: Maintain a small buffer list, append items as you iterate, and `yield` the buffer every N elements or at the end; this pattern is common when grouping work for database writes or network calls.
41. Concurrency models: Use threads for I/O-bound tasks, processes for CPU-bound work to bypass the GIL, and async IO for high-concurrency network-bound workloads using a single-threaded event loop.
42. `concurrent.futures`: Submit callables to a `ThreadPoolExecutor`/`ProcessPoolExecutor`, iterate over `as_completed` futures, and handle exceptions via `future.result()` wrapped in `try/except` (or by inspecting `future.exception()`).
43. Asyncio: An event loop runs coroutines defined with `async def` that `await` on I/O; it is ideal for many concurrent sockets/requests where threads would be too heavy or context-switching overhead is high.
44. Shared state and locks: The GIL does not protect your own data structures; use `threading.Lock`/`RLock`, queues, or higher-level thread-safe abstractions to guard shared mutable state and avoid deadlocks via consistent lock ordering.
45. Rate limiting: Track timestamps or tokens and block/sleep when a limit is reached, using locks or async primitives (`asyncio.Semaphore`, `BoundedSemaphore`) to enforce per-interval quotas across workers.
46. Logging: Configure hierarchical loggers with names (for example, `myapp.module`), handlers, and formatters; libraries should log to their own logger, and applications configure handlers/levels to control output without changing library code.
47. Testing: Write small, pure functions, inject dependencies, and avoid global state so you can test with `pytest`/`unittest`; use `unittest.mock` or monkeypatching to isolate external systems and side effects.
48. Type hints: Type hints improve readability, IDE support, and early error detection with tools like `mypy`, but they are optional at runtime; introduce them gradually around public APIs and critical modules.
49. Performance analysis: Use profiling (`cProfile`, line profilers) to locate hot spots, then address algorithmic complexity first before micro-optimizing Python-level overhead.
50. Beyond pure Python: When you still need more speed, reach for optimized libraries (NumPy/pandas), C-extensions/Cython/Numba, or move heavy computation into a separate service written in a faster language, trading simplicity for performance.

### Q1 - Names and objects

- A *name* in Python is just a key in a namespace (usually a dict) that refers
  to an object.
- Assignment like `b = a` copies the reference, not the underlying data.
- If two names point at the same **mutable** object and you mutate it, both see
  the change.
- Rebinding a name (for example, `a = []`) only changes which object that name
  points to; it does not affect other names.


```python
# Q1 - names and objects example
a = [1, 2]
b = a          # b refers to the same list as a
b.append(3)
print("a:", a)  # [1, 2, 3]
print("b:", b)  # [1, 2, 3]

c = a
a = [42]       # now a points at a new list; b and c still see the old list
print("a:", a)
print("b:", b)
print("c:", c)
```

### Q2 - Mutability and function arguments

- Python is "call by object reference" (also called "call by sharing").
- Function parameters are new local names that refer to the same objects as in
  the caller.
- Mutating a passed-in mutable object (list/dict/set) is visible to the caller.
- Rebinding the parameter name only affects the local name inside the function.


```python
# Q2 - mutability and arguments example
def append_in_place(lst, item):
    """Mutates the list in place; caller will see the change."""
    lst.append(item)


def rebind_parameter(lst):
    """Rebinds the local name only; caller will NOT see a new list."""
    lst = []  # shadows the caller's reference


nums = [1, 2]
append_in_place(nums, 3)
print("after append_in_place:", nums)

rebind_parameter(nums)
print("after rebind_parameter:", nums)  # unchanged
```

### Q3 - Loops vs comprehensions vs generators

- A plain `for` loop is most flexible (multiple statements, early `break`,
  complex control flow).
- A list comprehension builds a full list in memory; good when you *need* a
  list.
- A generator expression is lazy and yields one item at a time; great for large
  streams where you do not want to hold everything in memory.


```python
# Q3 - loops vs comprehensions vs generators example
nums = [1, 2, 3, 4]

squares_for = []
for x in nums:
    squares_for.append(x * 2)

squares_comp = [x * 2 for x in nums]
squares_gen = (x * 2 for x in nums)

print("for loop:", squares_for)
print("list comprehension:", squares_comp)
print("generator first two:", next(squares_gen), next(squares_gen))
```

### Q4 - `is` vs `==`

- `==` calls `__eq__` and checks **value equality**.
- `is` checks **object identity** (are these two names pointing at the *same*
  object in memory?).
- Use `is None` / `is not None` for `None` checks because `None` is a
  singleton.


```python
# Q4 - is vs == example
x = 256
y = 256
print("x == y:", x == y)  # same value
print("x is y:", x is y)  # CPython may intern small ints, but do not rely

a = [1, 2, 3]
b = [1, 2, 3]
print("a == b:", a == b)  # values equal
print("a is b:", a is b)  # different list objects
```

### Q5 - Truthiness

- `if x:` calls `bool(x)`.
- `bool(x)` prefers `x.__bool__()`, then falls back to `__len__()`, else
  defaults to `True`.
- Built-in falsy values include: `False`, `0`, `0.0`, `""`, `[]`, `{}`,
  `set()`, and `None`.
- Custom classes can control truthiness by defining `__bool__` or `__len__`.


```python
# Q5 - truthiness example
tests = [
    0,
    1,
    "",
    "hello",
    [],
    [1],
    None,
]

for value in tests:
    print(repr(value), "->", bool(value))
```

### Q6 - Built-in containers: list, tuple, set, dict

- **Lists**: mutable, ordered sequences. O(1) `append`/`pop` at end on average,
  O(n) insert/delete in the middle, O(1) indexing by position.
- **Tuples**: immutable sequences; good as dictionary keys and for fixed
  records. Same indexing semantics as lists but cannot be mutated.
- **Sets**: unordered collection of unique *hashable* elements implemented as a
  hash table. Average O(1) membership tests, add, and remove.
- **Dicts**: mapping from hashable keys to values, also backed by a hash table
  with average O(1) lookup/insert/delete.


```python
# Q6 - containers example
numbers = [1, 2, 3]
numbers.append(4)
coords = (10, 20)
unique = {1, 2, 2, 3}
mapping = {"a": 1, "b": 2}

print("list:", numbers)
print("tuple:", coords)
print("set:", unique)
print("dict:", mapping)
```

### Q7 - Hashing and hashable objects

- An object is **hashable** if its hash value is stable for its lifetime and it
  can be compared for equality. In Python that means implementing `__hash__`
  and `__eq__` consistently.
- In hash-based containers (dict/set), if `a == b` then you must have
  `hash(a) == hash(b)` or lookups will misbehave.
- Mutable objects generally should **not** be hashable because changing the
  fields used in the hash breaks membership and lookup.


```python
# Q7 - hashing example
class Point:
    def __init__(self, x: int, y: int) -> None:
        self.x = x
        self.y = y

    def __eq__(self, other: object) -> bool:
        return isinstance(other, Point) and (self.x, self.y) == (other.x, other.y)

    def __hash__(self) -> int:  # hash depends only on immutable pair (x, y)
        return hash((self.x, self.y))

seen = {Point(1, 2)}
print(Point(1, 2) in seen)
```

### Q8 - Shallow vs deep copying

- A **shallow copy** of a container (for example, `list(x)` or `x.copy()`) creates
  a new outer container but reuses references to the same inner objects.
- A **deep copy** (via `copy.deepcopy`) recursively copies nested structures so
  that changes to inner objects do not affect the original.
- Shallow copy is usually faster and often what you want; deep copy is reserved
  for when you truly need full independence.


```python
# Q8 - shallow vs deep copy example
import copy

original = [[1], [2]]
shallow = list(original)
deep = copy.deepcopy(original)

original[0].append(99)
print("original:", original)
print("shallow:", shallow)  # sees inner mutation
print("deep:", deep)        # isolated
```

### Q9 - collections utilities

- `deque` from `collections` is ideal for queues and BFS: fast O(1)
  appends/pops from both ends.
- `defaultdict` simplifies counting and grouping by auto-creating missing
  entries.
- `Counter` is a specialized dict for counting hashable items.


```python
# Q9 - collections utilities example
from collections import deque, defaultdict, Counter

q = deque([1, 2])
q.appendleft(0)

dd = defaultdict(int)
for ch in "banana":
    dd[ch] += 1

cnt = Counter("banana")

print("deque:", list(q))
print("defaultdict counts:", dict(dd))
print("Counter counts:", dict(cnt))
```

### Q10 - Iterators and generators

- An **iterator** implements `__iter__` (returning itself) and `__next__`, which
  either yields the next value or raises `StopIteration`.
- A **generator function** is defined with `def` and uses `yield`; calling it
  returns a generator object that implements the iterator protocol.
- Generators are memory-efficient for large or infinite sequences since they
  produce values lazily.


```python
# Q10 - iterator and generator example
def count_up_to(n: int):
    """A simple generator that yields numbers from 1 to n."""

    current = 1
    while current <= n:
        yield current
        current += 1

for value in count_up_to(3):
    print(value)
```

### Q11 - Attribute lookup

- Python looks up attributes in this order: instance `__dict__` → class → base
  classes (following MRO).
- Class attributes are shared across all instances.
- Mutable class attributes (like lists/dicts) are a common bug: all instances see
  the same object.


```python
# Q11 - attribute lookup example
class Counter:
    count = 0  # class attribute

    def __init__(self, name):
        self.name = name  # instance attribute
        Counter.count += 1


c1 = Counter("first")
c2 = Counter("second")
print("Total counters:", Counter.count)  # 2
print("c1.count:", c1.count)  # 2 (found via class)
print("c1.name:", c1.name)  # "first" (instance attribute)


# Mutable class attribute bug
class BuggyList:
    items = []  # shared across all instances!

    def add(self, x):
        self.items.append(x)


b1 = BuggyList()
b2 = BuggyList()
b1.add(1)
b2.add(2)
print("b1.items:", b1.items)  # [1, 2] - both see the same list!
print("b2.items:", b2.items)  # [1, 2]
```

### Q12 - Methods

- **Instance method**: Takes `self`, operates on instance data.
- **`@classmethod`**: Takes `cls`, good for alternative constructors or factory
  methods.
- **`@staticmethod`**: Takes neither `self` nor `cls`, just a utility function
  namespaced in the class.


```python
# Q12 - methods example
from datetime import datetime


class Person:
    def __init__(self, name, birth_year):
        self.name = name
        self.birth_year = birth_year

    def age(self):
        """Instance method - uses self."""
        return datetime.now().year - self.birth_year

    @classmethod
    def from_birth_date(cls, name, birth_date_str):
        """Class method - alternative constructor."""
        year = int(birth_date_str.split("-")[0])
        return cls(name, year)

    @staticmethod
    def is_adult(age):
        """Static method - utility function."""
        return age >= 18


p1 = Person("Alice", 1990)
print("Age:", p1.age())

p2 = Person.from_birth_date("Bob", "1985-06-15")
print("Bob's age:", p2.age())

print("Is 20 adult?", Person.is_adult(20))
```

### Q13 - Dunder methods

- `__repr__`: Unambiguous representation for debugging (should ideally be valid
  Python).
- `__str__`: User-friendly string for `print()`.
- `__eq__`, `__lt__`, etc.: Enable comparisons and sorting.
- `__len__`, `__iter__`, `__contains__`: Make objects work with `len()`, `for`,
  and `in`.


```python
# Q13 - dunder methods example
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __lt__(self, other):
        return (self.x, self.y) < (other.x, other.y)


p1 = Point(1, 2)
p2 = Point(1, 2)
p3 = Point(2, 3)

print(repr(p1))  # Point(1, 2)
print(str(p1))  # (1, 2)
print(p1 == p2)  # True
print(p1 < p3)  # True
print(sorted([p3, p1]))  # [Point(1, 2), Point(2, 3)]
```

### Q15 - Context managers

- Context managers implement `__enter__` and `__exit__`.
- `with` statement calls `__enter__` before the block and `__exit__` after,
  ensuring cleanup even on exceptions.
- Common use: files, locks, database connections, timers.


```python
# Q15 - context manager example
class Timer:
    def __enter__(self):
        import time

        self.start = time.time()
        print("Timer started")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        import time

        elapsed = time.time() - self.start
        print(f"Timer stopped: {elapsed:.4f}s")
        return False  # don't suppress exceptions


with Timer():
    total = sum(range(1000000))
```

### Q16 - Exception handling

- `try/except/else/finally`: `try` contains risky code, `except` handles errors,
  `else` runs if no exception, `finally` always runs (cleanup).
- Exceptions are more Pythonic than return codes because they separate normal
  flow from error handling.


```python
# Q16 - exception handling example
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None
    except TypeError:
        print("Invalid types")
        return None
    else:
        print("Division successful")
        return result
    finally:
        print("Cleanup always runs")


print(divide(10, 2))
print(divide(10, 0))
```

### Q17 - Memory management

- CPython uses **reference counting**: each object has a refcount; when it hits
  zero, memory is freed.
- **Cyclic GC**: Detects and collects reference cycles (e.g., `a.next = b`,
  `b.prev = a`).
- `del x` removes the name binding; the object is freed when refcount → 0.


```python
# Q17 - memory management example
import sys


class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


a = Node(1)
print("Refcount of a:", sys.getrefcount(a) - 1)  # -1 for getrefcount's own ref

b = a  # another reference
print("Refcount after b=a:", sys.getrefcount(a) - 1)

del b
print("Refcount after del b:", sys.getrefcount(a) - 1)
```

### Q18 - GIL and concurrency

- **GIL (Global Interpreter Lock)**: Only one thread executes Python bytecode at
  a time in CPython.
- **CPU-bound**: Threads don't help; use `multiprocessing` or async for I/O.
- **I/O-bound**: Threads work well because GIL is released during I/O waits.


```python
# Q18 - GIL example (conceptual)
import threading
import time


def cpu_bound_task():
    total = 0
    for i in range(5_000_000):
        total += i
    return total


# CPU-bound with threads: won't speed up due to GIL
start = time.time()
t1 = threading.Thread(target=cpu_bound_task)
t2 = threading.Thread(target=cpu_bound_task)
t1.start()
t2.start()
t1.join()
t2.join()
print(f"Threaded CPU-bound: {time.time() - start:.2f}s")

# For CPU-bound, use multiprocessing instead (not shown here)
```

### Q19 - Deadlocks and race conditions

- **Deadlock**: Thread A holds lock1 and waits for lock2; thread B holds lock2
  and waits for lock1.
- **Avoidance**: Consistent lock ordering, use timeouts, or higher-level
  primitives like queues.
- **Race condition**: Multiple threads access shared mutable state without proper
  synchronization.


```python
# Q19 - deadlock example (conceptual - don't run in production!)
import threading

lock1 = threading.Lock()
lock2 = threading.Lock()


def task_a():
    with lock1:
        print("A acquired lock1")
        time.sleep(0.1)
        with lock2:
            print("A acquired lock2")


def task_b():
    with lock2:
        print("B acquired lock2")
        time.sleep(0.1)
        with lock1:
            print("B acquired lock1")


# If both run simultaneously, they can deadlock
# (Commented out to avoid hanging)
# threading.Thread(target=task_a).start()
# threading.Thread(target=task_b).start()

print("Deadlock example (not executed)")
```

### Q21 - Mutable default arguments

- Default arguments are evaluated **once** at function definition time.
- Mutable defaults (list/dict) are reused across calls, causing bugs.
- **Fix**: Use `None` as default and initialize inside the function.


```python
# Q21 - mutable default arguments example
def buggy_append(item, lst=[]):
    """Bug: default list is shared across calls."""
    lst.append(item)
    return lst


print(buggy_append(1))  # [1]
print(buggy_append(2))  # [1, 2] - unexpected!


def fixed_append(item, lst=None):
    """Fixed: create new list each time."""
    if lst is None:
        lst = []
    lst.append(item)
    return lst


print(fixed_append(1))  # [1]
print(fixed_append(2))  # [2] - correct!
```

### Q22 - Function signatures

- **Positional-only** (`/`): Arguments before `/` can only be passed
  positionally.
- **Keyword-only** (`*`): Arguments after `*` must be passed by name.
- `*args`, `**kwargs`: Capture variable positional and keyword arguments.

### Q24 - Decorators

- A decorator is a function that takes a function and returns a new function.
- Used for cross-cutting concerns: logging, timing, authentication, caching.
- Use `functools.wraps` to preserve metadata (`__name__`, `__doc__`).


```python
# Q24 - decorators example
import functools
import time


def timing_decorator(func):
    """Decorator that times function execution."""

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        elapsed = time.time() - start
        print(f"{func.__name__} took {elapsed:.4f}s")
        return result

    return wrapper


@timing_decorator
def slow_computation(n):
    """Compute sum of squares."""
    return sum(i**2 for i in range(n))


result = slow_computation(100000)
print("Result:", result)
```

### Q25 - Caching

- `functools.lru_cache` memoizes function results based on arguments.
- **Requirements**: Arguments must be hashable, function must be pure
  (deterministic).
- **Trade-off**: Memory for speed.
- **Pitfall**: Mutable arguments or external state breaks caching.


```python
# Q25 - caching example
from functools import lru_cache


@lru_cache(maxsize=128)
def fibonacci(n):
    """Cached Fibonacci - much faster for repeated calls."""
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(30))  # Fast due to caching
print(fibonacci.cache_info())  # Show cache statistics
```

### Q27 - Contextlib

- `@contextlib.contextmanager` lets you write a generator-based context manager.
- Yield once: code before `yield` is `__enter__`, code after is `__exit__`.
- More concise than writing a full class for simple cases.


```python
# Q27 - contextlib example
from contextlib import contextmanager


@contextmanager
def temporary_value(var_dict, key, temp_value):
    """Temporarily set a dict value, then restore."""
    old_value = var_dict.get(key)
    var_dict[key] = temp_value
    try:
        yield
    finally:
        if old_value is None:
            var_dict.pop(key, None)
        else:
            var_dict[key] = old_value


config = {"debug": False}
print("Before:", config)

with temporary_value(config, "debug", True):
    print("Inside context:", config)

print("After:", config)
```

### Q28 - Properties

- Use plain attributes for simple data.
- Use `@property` for computed or validated access with attribute-like syntax.
- **Pitfall**: Heavy or side-effectful properties surprise callers and hurt
  performance.


```python
# Q28 - properties example
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        """Getter for radius."""
        return self._radius

    @radius.setter
    def radius(self, value):
        """Setter with validation."""
        if value < 0:
            raise ValueError("Radius must be non-negative")
        self._radius = value

    @property
    def area(self):
        """Computed property."""
        import math

        return math.pi * self._radius**2


c = Circle(5)
print("Radius:", c.radius)
print("Area:", c.area)
c.radius = 10
print("New area:", c.area)
```

### Q29 - Descriptors

- Descriptors implement `__get__`, `__set__`, `__delete__` to control attribute
  access.
- `@property` is implemented using descriptors.
- Useful for reusable validation, lazy loading, or computed attributes.


```python
# Q29 - descriptors example
class NonNegative:
    """Descriptor that enforces non-negative values."""

    def __init__(self, name):
        self.name = name

    def __get__(self, obj, objtype=None):
        if obj is None:
            return self
        return obj.__dict__.get(self.name, 0)

    def __set__(self, obj, value):
        if value < 0:
            raise ValueError(f"{self.name} must be non-negative")
        obj.__dict__[self.name] = value


class Rectangle:
    width = NonNegative("width")
    height = NonNegative("height")

    def __init__(self, width, height):
        self.width = width
        self.height = height


r = Rectangle(10, 20)
print("Width:", r.width)
# r.width = -5  # Raises ValueError
```

### Q31 - Dynamic attribute access

- `__getattr__`: Called only when attribute is **not found** normally.
- `__getattribute__`: Called for **every** attribute access.
- Use `__getattr__` for proxies, lazy loading; avoid infinite recursion in
  `__getattribute__` by calling `super().__getattribute__`.


```python
# Q31 - dynamic attribute access example
class LazyLoader:
    def __init__(self):
        self.data = {"x": 10, "y": 20}

    def __getattr__(self, name):
        """Called when attribute not found normally."""
        print(f"__getattr__ called for {name}")
        if name in self.data:
            return self.data[name]
        raise AttributeError(f"No attribute {name}")


obj = LazyLoader()
print(obj.x)  # Calls __getattr__
# print(obj.z)  # Raises AttributeError


class LoggingProxy:
    def __init__(self, target):
        object.__setattr__(self, "_target", target)

    def __getattribute__(self, name):
        """Intercepts all attribute access."""
        if name == "_target":
            return object.__getattribute__(self, name)
        print(f"Accessing {name}")
        target = object.__getattribute__(self, "_target")
        return getattr(target, name)


class Data:
    value = 42


proxy = LoggingProxy(Data())
print(proxy.value)
```

### Q34 - Rich comparisons

- Implement `__eq__` and one ordering method (e.g., `__lt__`).
- `functools.total_ordering` auto-generates the rest.
- Must define a consistent total order.


```python
# Q34 - rich comparisons example
from functools import total_ordering


@total_ordering
class Version:
    def __init__(self, major, minor):
        self.major = major
        self.minor = minor

    def __eq__(self, other):
        return (self.major, self.minor) == (other.major, other.minor)

    def __lt__(self, other):
        return (self.major, self.minor) < (other.major, other.minor)

    def __repr__(self):
        return f"Version({self.major}.{self.minor})"


v1 = Version(1, 2)
v2 = Version(1, 3)
v3 = Version(2, 0)

print(v1 < v2)  # True
print(v2 <= v3)  # True (auto-generated)
print(v3 > v1)  # True (auto-generated)
print(sorted([v3, v1, v2]))
```

### Q35 - Hash and equality for custom types

- For dict/set keys, implement `__hash__` and `__eq__`.
- **Requirement**: If `a == b`, then `hash(a) == hash(b)`.
- Only hash **immutable** fields; mutating hashed objects breaks dicts/sets.


```python
# Q35 - hash and equality example
class ImmutablePoint:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    @property
    def x(self):
        return self._x

    @property
    def y(self):
        return self._y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __hash__(self):
        return hash((self.x, self.y))

    def __repr__(self):
        return f"ImmutablePoint({self.x}, {self.y})"


p1 = ImmutablePoint(1, 2)
p2 = ImmutablePoint(1, 2)
p3 = ImmutablePoint(3, 4)

print(p1 == p2)  # True
print(hash(p1) == hash(p2))  # True

# Can use as dict key
points_map = {p1: "origin", p3: "other"}
print(points_map[p2])  # "origin" (p2 == p1)
```

### Q40 - Batching

- Yield items in fixed-size batches for downstream consumers.
- Useful for database writes, network calls, or parallel processing.


```python
# Q40 - batching example
def batch_iterator(iterable, batch_size):
    """Yield items in batches of batch_size."""
    batch = []
    for item in iterable:
        batch.append(item)
        if len(batch) == batch_size:
            yield batch
            batch = []
    if batch:  # Yield remaining items
        yield batch


data = range(10)
for batch in batch_iterator(data, 3):
    print("Batch:", batch)
```

### Q41 - Concurrency models

- **Threads**: Good for I/O-bound tasks; GIL limits CPU-bound parallelism.
- **Processes**: Bypass GIL, good for CPU-bound work; higher overhead.
- **Async IO**: Single-threaded event loop, ideal for high-concurrency network
  servers.


```python
# Q41 - concurrency models example
import asyncio
import threading
from concurrent.futures import ThreadPoolExecutor, ProcessPoolExecutor


def io_task(n):
    """Simulated I/O-bound task."""
    import time

    time.sleep(0.1)
    return n * 2


# Threads for I/O-bound
with ThreadPoolExecutor(max_workers=3) as executor:
    results = list(executor.map(io_task, range(5)))
    print("Thread results:", results)


# Async for I/O-bound
async def async_io_task(n):
    await asyncio.sleep(0.1)
    return n * 2


async def run_async():
    tasks = [async_io_task(i) for i in range(5)]
    results = await asyncio.gather(*tasks)
    print("Async results:", results)


asyncio.run(run_async())
```

### Q42 - `concurrent.futures`

- `ThreadPoolExecutor` / `ProcessPoolExecutor` provide a high-level interface for
  parallelism.
- Submit tasks, get futures, handle exceptions via `future.result()`.


```python
# Q42 - concurrent.futures example
from concurrent.futures import ThreadPoolExecutor, as_completed


def task(n):
    if n == 3:
        raise ValueError("Task 3 failed")
    return n * n


with ThreadPoolExecutor(max_workers=4) as executor:
    futures = {executor.submit(task, i): i for i in range(5)}

    for future in as_completed(futures):
        i = futures[future]
        try:
            result = future.result()
            print(f"Task {i} result: {result}")
        except Exception as e:
            print(f"Task {i} failed: {e}")
```

### Q44 - Shared state and locks

```python
# Q44 - shared state and locks example
import threading

counter = 0
lock = threading.Lock()


def increment():
    global counter
    for _ in range(100000):
        with lock:
            counter += 1


threads = [threading.Thread(target=increment) for _ in range(5)]
for t in threads:
    t.start()
for t in threads:
    t.join()

print("Final counter:", counter)  # Should be 500000
```

### Q45 - Rate limiting / throttling

- Track timestamps or use a token bucket.
- Use locks or async primitives (`asyncio.Semaphore`) to enforce limits.


```python
# Q45 - rate limiting example
import asyncio
import time


class RateLimiter:
    def __init__(self, rate, per):
        """Allow 'rate' calls per 'per' seconds."""
        self.rate = rate
        self.per = per
        self.allowance = rate
        self.last_check = time.time()
        self.lock = asyncio.Lock()

    async def acquire(self):
        async with self.lock:
            current = time.time()
            elapsed = current - self.last_check
            self.last_check = current
            self.allowance += elapsed * (self.rate / self.per)
            if self.allowance > self.rate:
                self.allowance = self.rate

            if self.allowance < 1:
                sleep_time = (1 - self.allowance) * (self.per / self.rate)
                await asyncio.sleep(sleep_time)
                self.allowance = 0
            else:
                self.allowance -= 1


async def make_request(limiter, i):
    await limiter.acquire()
    print(f"Request {i} at {time.time():.2f}")


async def test_rate_limiter():
    limiter = RateLimiter(rate=2, per=1)  # 2 requests per second
    tasks = [make_request(limiter, i) for i in range(5)]
    await asyncio.gather(*tasks)


asyncio.run(test_rate_limiter())
```

### Q50 - Beyond pure Python

- **Optimized libraries**: NumPy, pandas for numerical/data work.

- **Cython**: Python-like syntax compiled to C.
- **Numba**: JIT compiler for numerical code.

```python
# Q50 - beyond pure Python example
import numpy as np

# Pure Python
def sum_squares_python(n):
    return sum(i**2 for i in range(n))


# NumPy (much faster for large n)
def sum_squares_numpy(n):
    return np.sum(np.arange(n) ** 2)


import timeit

n = 100000
python_time = timeit.timeit(lambda: sum_squares_python(n), number=10)
numpy_time = timeit.timeit(lambda: sum_squares_numpy(n), number=10)

print(f"Python: {python_time:.4f}s")
print(f"NumPy: {numpy_time:.4f}s")
print(f"Speedup: {python_time / numpy_time:.1f}x")
```
