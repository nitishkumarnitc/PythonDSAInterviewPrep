### Question 1 – Names and Objects

**Question:**

Explain how Python names and objects work. What happens when you assign `b = a`?
How does this differ from languages with value semantics?

**Answer:**

A *name* in Python is a reference to an object, not a container holding a value.
Assignment like `b = a` copies the reference, not the underlying data. If two names
point to the same mutable object and you mutate it, both see the change. Rebinding
a name (e.g., `a = []`) only changes which object that name points to; it doesn't
affect other names.

**Example Code:**


```
# Names and objects demonstration
a = [1, 2]
b = a          # b refers to the same list as a
b.append(3)
print("a:", a)  # [1, 2, 3] - both see the change
print("b:", b)  # [1, 2, 3]

c = a
a = [42]       # now a points at a new list; b and c still see the old list
print("After rebinding a:")
print("a:", a)  # [42]
print("b:", b)  # [1, 2, 3]
print("c:", c)  # [1, 2, 3]
```

### Question 2 – Mutability and Function Arguments

**Question:**

Explain Python's argument passing mechanism. What happens when you pass a list to
a function and modify it? What about rebinding the parameter?

**Answer:**

Python uses "call by object reference" (also called "call by sharing"). Function
parameters are new local names that refer to the same objects as in the caller.
Mutating a passed-in mutable object (list/dict/set) is visible to the caller.
Rebinding the parameter name only affects the local name inside the function and
doesn't affect the caller's reference.

**Example Code:**


```
# Mutability and arguments demonstration
def append_in_place(lst, item):
    lst.append(item)  # Mutates the original list

def rebind_parameter(lst, item):
    lst = [item]  # Only rebinds local name, doesn't affect caller
    return lst

original = [1, 2]
append_in_place(original, 3)
print("After append_in_place:", original)  # [1, 2, 3]

result = rebind_parameter(original, 99)
print("After rebind_parameter:")
print("  original:", original)  # [1, 2, 3] - unchanged
print("  result:", result)      # [99]
```

### Question 3 – Loops vs Comprehensions vs Generators

**Question:**

Compare for-loops, list comprehensions, and generator expressions. When should
you use each?

**Answer:**

For-loops are explicit and flexible, good for complex logic or side effects.
List comprehensions are concise and fast but allocate the entire result list
in memory immediately. Generator expressions are lazy and memory-efficient,
ideal for large datasets or streaming data where you process items one at a time.

**Example Code:**


```
# Loops vs comprehensions vs generators
import sys

# For-loop (explicit, flexible)
result_loop = []
for i in range(5):
    result_loop.append(i ** 2)
print("For-loop:", result_loop)

# List comprehension (concise, eager)
result_list = [i ** 2 for i in range(5)]
print("List comp:", result_list)

# Generator expression (lazy, memory-efficient)
result_gen = (i ** 2 for i in range(5))
print("Generator:", result_gen)
print("Generator values:", list(result_gen))

# Memory comparison
large_list = [x for x in range(1000000)]
large_gen = (x for x in range(1000000))
print(f"List size: {sys.getsizeof(large_list):,} bytes")
print(f"Generator size: {sys.getsizeof(large_gen):,} bytes")
```

### Question 4 – `is` vs `==`

**Question:**

What's the difference between `is` and `==`? When should you use each?

**Answer:**

`==` tests value equality via the `__eq__` method. `is` tests identity (whether
two names refer to the exact same object in memory). Use `is None` to check for
the singleton `None`. For most comparisons, use `==`. Use `is` only for singletons
like `None`, `True`, `False`, or when you specifically need identity checks.

**Example Code:**


```
# is vs == demonstration
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)  # True (same values)
print("a is b:", a is b)  # False (different objects)
print("a is c:", a is c)  # True (same object)

# Singleton check
x = None
print("x is None:", x is None)  # Correct way
print("x == None:", x == None)  # Works but not idiomatic

# Small integer caching (CPython implementation detail)
m = 256
n = 256
print("256 is 256:", m is n)  # True (cached)

p = 257
q = 257
print("257 is 257:", p is q)  # May be False (not guaranteed to be cached)

```

### Question 5 – Truthiness

**Question:**

How does Python determine truthiness? What values are falsy? How can custom classes control their truthiness?

**Answer:**

When you write `if x:`, Python calls `bool(x)`. The `bool()` function first tries `x.__bool__()`, then falls back to `__len__()` (where 0 is falsy), and defaults to `True` if neither is defined. Built-in falsy values include: `False`, `0`, `0.0`, `""` (empty string), `[]` (empty list), `{}` (empty dict), `set()` (empty set), and `None`. Custom classes can define `__bool__` or `__len__` to control their truthiness.

**Example Code:**


```
# Truthiness demonstration
tests = [
    False,
    0,
    0.0,
    "",
    [],
    {},
    set(),
    None,
    1,
    "hello",
    [1, 2],
]

print("Truthiness tests:")
for value in tests:
    print(f"{repr(value):15} -> {bool(value)}")

# Custom class with __bool__
class AlwaysFalsy:
    def __bool__(self):
        return False

obj = AlwaysFalsy()
print(f"\nAlwaysFalsy() -> {bool(obj)}")  # False
```

### Question 6 – Built-in Containers: list, tuple, set, dict

**Question:**

Compare the four main built-in container types in Python. What are their key characteristics and performance properties?

**Answer:**

**Lists** are mutable, ordered sequences with O(1) append/pop at the end (amortized), O(n) insert/delete in the middle, and O(1) indexing. **Tuples** are immutable sequences, good as dictionary keys and for fixed records; same indexing as lists but cannot be mutated. **Sets** are unordered collections of unique hashable elements implemented as hash tables, with average O(1) membership tests, add, and remove. **Dicts** map hashable keys to values, also backed by hash tables with average O(1) lookup/insert/delete.

**Example Code:**


```
# Container types demonstration
# List - mutable, ordered
numbers = [1, 2, 3]
numbers.append(4)
numbers[0] = 10
print("List:", numbers)  # [10, 2, 3, 4]

# Tuple - immutable, ordered
coords = (10, 20)
# coords[0] = 5  # Would raise TypeError
print("Tuple:", coords)

# Set - unique elements, unordered
unique = {1, 2, 2, 3, 3, 3}
print("Set:", unique)  # {1, 2, 3}
print("2 in set:", 2 in unique)  # O(1) membership

# Dict - key-value mapping
mapping = {"a": 1, "b": 2}
mapping["c"] = 3
print("Dict:", mapping)
print("Lookup 'b':", mapping["b"])  # O(1) lookup
```

### Question 7 – Hashing and Hashable Objects

**Question:**

What makes an object hashable? Why is the relationship between `__hash__` and `__eq__` important?

**Answer:**

An object is hashable if it has a stable `__hash__` value for its lifetime and implements `__eq__` consistently. The critical invariant is: if `a == b`, then `hash(a) == hash(b)`. This is required for hash-based containers (dict/set) to work correctly. Mutable objects generally should not be hashable because changing fields used in the hash breaks membership and lookup. Immutable built-ins (int, str, tuple of hashables) are hashable; mutable ones (list, dict, set) are not.

**Example Code:**


```
# Hashing demonstration
class Point:
    def __init__(self, x: int, y: int):
        self.x = x
        self.y = y

    def __eq__(self, other):
        if not isinstance(other, Point):
            return False
        return (self.x, self.y) == (other.x, other.y)

    def __hash__(self):
        # Hash based on immutable tuple
        return hash((self.x, self.y))

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

# Use Point as dict key and in set
p1 = Point(1, 2)
p2 = Point(1, 2)
p3 = Point(3, 4)

seen = {p1, p3}
print("Set of points:", seen)
print("p2 in seen:", p2 in seen)  # True (p2 == p1)

point_data = {p1: "origin area", p3: "far point"}
print("Dict lookup:", point_data[p2])  # Works because p2 == p1
```

### Question 8 – Shallow vs Deep Copying

**Question:**

What's the difference between shallow and deep copying? When would you use each?

**Answer:**

A **shallow copy** creates a new outer container but reuses references to the same inner objects. You can create one with `list(x)`, `x.copy()`, or `copy.copy(x)`. A **deep copy** (via `copy.deepcopy()`) recursively copies nested structures so changes to inner objects don't affect the original. Shallow copy is faster and usually sufficient; deep copy is needed when you require full independence of nested mutable structures.

**Example Code:**


```
# Shallow vs deep copy demonstration
import copy

original = [[1, 2], [3, 4]]
shallow = list(original)  # or original.copy()
deep = copy.deepcopy(original)

print("Before mutation:")
print("original:", original)
print("shallow:", shallow)
print("deep:", deep)

# Mutate inner list
original[0].append(99)

print("\nAfter original[0].append(99):")
print("original:", original)  # [[1, 2, 99], [3, 4]]
print("shallow:", shallow)    # [[1, 2, 99], [3, 4]] - sees the change!
print("deep:", deep)          # [[1, 2], [3, 4]] - isolated
```

### Question 9 – collections Utilities

**Question:**

Describe useful classes from the `collections` module and their use cases.

**Answer:**

**`deque`** (double-ended queue) provides O(1) appends/pops from both ends, ideal for queues and BFS. **`defaultdict`** auto-initializes missing keys with a factory function, simplifying counting and grouping. **`Counter`** is a specialized dict for counting hashable items with convenient methods like `most_common()`. **`namedtuple`** creates lightweight immutable records with named fields. **`OrderedDict`** (less needed in Python 3.7+) maintains insertion order explicitly.

**Example Code:**


```
# collections utilities demonstration
from collections import deque, defaultdict, Counter

# deque - efficient queue
q = deque([1, 2, 3])
q.appendleft(0)  # O(1) prepend
q.append(4)      # O(1) append
print("deque:", list(q))  # [0, 1, 2, 3, 4]

# defaultdict - auto-initialize missing keys
word_lengths = defaultdict(list)
for word in ["cat", "dog", "bird", "ant"]:
    word_lengths[len(word)].append(word)
print("defaultdict:", dict(word_lengths))  # {3: ['cat', 'dog', 'ant'], 4: ['bird']}

# Counter - counting
letter_counts = Counter("banana")
print("Counter:", letter_counts)  # Counter({'a': 3, 'n': 2, 'b': 1})
print("Most common:", letter_counts.most_common(2))  # [('a', 3), ('n', 2)]
```

### Question 10 – Iterators and Generators

**Question:**

Explain the iterator protocol and how generators implement it. Why are generators useful?

**Answer:**

An **iterator** implements `__iter__()` (returns self) and `__next__()` (returns next value or raises `StopIteration`). A **generator function** uses `yield` to produce values lazily; calling it returns a generator object that implements the iterator protocol automatically. Generators are memory-efficient for large or infinite sequences since they produce values on-demand rather than storing everything in memory. They're ideal for streaming data, pipelines, and processing large files.

**Example Code:**


```
# Iterator and generator demonstration
def count_up_to(n):
    """Generator that yields numbers from 1 to n."""
    current = 1
    while current <= n:
        yield current
        current += 1

# Generator is lazy - doesn't compute all values upfront
gen = count_up_to(5)
print("First value:", next(gen))  # 1
print("Second value:", next(gen))  # 2

# Can iterate over remaining values
print("Remaining:", list(gen))  # [3, 4, 5]

# Generator expression (like list comprehension but lazy)
squares_gen = (x**2 for x in range(1000000))
print("First 5 squares:", [next(squares_gen) for _ in range(5)])

# Memory comparison
import sys
list_comp = [x for x in range(10000)]
gen_expr = (x for x in range(10000))
print(f"\nList size: {sys.getsizeof(list_comp)} bytes")
print(f"Generator size: {sys.getsizeof(gen_expr)} bytes")
```

### Question 11 – Attribute Lookup

**Question:**

Explain Python's attribute lookup order. What are the pitfalls of mutable class attributes?

**Answer:**

Python looks up attributes in this order: instance `__dict__` → class → base classes (following MRO - Method Resolution Order). Class attributes are shared across all instances. A common bug is using mutable class attributes (like lists or dicts) as defaults - all instances see and modify the same object. Always initialize mutable instance attributes in `__init__`.

**Example Code:**


```
# Attribute lookup demonstration
class Counter:
    count = 0  # Class attribute - shared across all instances

    def __init__(self, name):
        self.name = name  # Instance attribute
        Counter.count += 1

c1 = Counter("first")
c2 = Counter("second")
print("Total counters:", Counter.count)  # 2
print("c1.count:", c1.count)  # 2 (found via class)
print("c1.name:", c1.name)  # "first" (instance attribute)

# Mutable class attribute bug
class BuggyList:
    items = []  # DANGER: shared across all instances!

    def add(self, x):
        self.items.append(x)

b1 = BuggyList()
b2 = BuggyList()
b1.add(1)
b2.add(2)
print("\nBuggy behavior:")
print("b1.items:", b1.items)  # [1, 2] - both see the same list!
print("b2.items:", b2.items)  # [1, 2]

# Fixed version
class FixedList:
    def __init__(self):
        self.items = []  # Each instance gets its own list

    def add(self, x):
        self.items.append(x)

f1 = FixedList()
f2 = FixedList()
f1.add(1)
f2.add(2)
print("\nFixed behavior:")
print("f1.items:", f1.items)  # [1]
print("f2.items:", f2.items)  # [2]
```

### Question 12 – Methods: Instance, Class, and Static

**Question:**

Compare instance methods, class methods, and static methods. When should you use each?

**Answer:**

**Instance methods** take `self` and operate on instance data - the most common type. **Class methods** (decorated with `@classmethod`) take `cls` as the first parameter and are good for alternative constructors or factory methods that need access to the class. **Static methods** (decorated with `@staticmethod`) take neither `self` nor `cls` - they're utility functions namespaced within the class for organizational purposes.

**Example Code:**


```
# Methods demonstration
from datetime import datetime

class Person:
    def __init__(self, name, birth_year):
        self.name = name
        self.birth_year = birth_year

    def age(self):
        """Instance method - uses self."""
        return datetime.now().year - self.birth_year

    @classmethod
    def from_birth_date(cls, name, birth_date_str):
        """Class method - alternative constructor."""
        year = int(birth_date_str.split("-")[0])
        return cls(name, year)

    @staticmethod
    def is_adult(age):
        """Static method - utility function."""
        return age >= 18

# Instance method
p1 = Person("Alice", 1990)
print("Alice's age:", p1.age())

# Class method as alternative constructor
p2 = Person.from_birth_date("Bob", "1985-06-15")
print("Bob's age:", p2.age())

# Static method
print("Is 20 adult?", Person.is_adult(20))  # True
print("Is 15 adult?", Person.is_adult(15))  # False
```

### Question 13 – Dunder Methods

**Question:**

What are dunder (double underscore) methods? Give examples of commonly used ones.

**Answer:**

Dunder methods (also called magic methods or special methods) allow custom classes to integrate with Python's built-in operations. `__repr__` provides an unambiguous representation for debugging (ideally valid Python code). `__str__` gives a user-friendly string for `print()`. Comparison dunders (`__eq__`, `__lt__`, etc.) enable comparisons and sorting. `__len__`, `__iter__`, `__contains__` make objects work with `len()`, `for` loops, and `in` checks.

**Example Code:**


```
# Dunder methods demonstration
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        """Unambiguous representation for debugging."""
        return f"Point({self.x}, {self.y})"

    def __str__(self):
        """User-friendly string."""
        return f"({self.x}, {self.y})"

    def __eq__(self, other):
        """Equality comparison."""
        if not isinstance(other, Point):
            return False
        return self.x == other.x and self.y == other.y

    def __lt__(self, other):
        """Less-than for sorting."""
        return (self.x, self.y) < (other.x, other.y)

    def __len__(self):
        """Distance from origin (for demonstration)."""
        return int((self.x**2 + self.y**2)**0.5)

p1 = Point(1, 2)
p2 = Point(1, 2)
p3 = Point(2, 3)

print("repr:", repr(p1))  # Point(1, 2)
print("str:", str(p1))    # (1, 2)
print("p1 == p2:", p1 == p2)  # True
print("p1 < p3:", p1 < p3)    # True
print("sorted:", sorted([p3, p1, p2]))  # [Point(1, 2), Point(1, 2), Point(2, 3)]
print("len(p3):", len(p3))  # 3

```

### Question 14 – Modules and Packages

**Question:**

What's the difference between a module and a package? How does Python's import mechanism work?

**Answer:**

A **module** is a single `.py` file. A **package** is a directory containing an `__init__.py` file (or just a directory in Python 3.3+ with namespace packages). **Example Code:**


```
# Modules and packages demonstration
import sys
import math

# Show some paths Python searches
print("Sample sys.path entries:", sys.path[:3])

# Modules are cached in sys.modules
print("\nmath in sys.modules:", "math" in sys.modules)  # True

# Importing again returns the cached module
import math as math2
print("Same object:", math is math2)  # True
```

### Question 15 – Context Managers

**Question:**

What are context managers? How do they work and when should you use them?

**Answer:**

Context managers implement `__enter__` and `__exit__` methods. The `with` statement calls `__enter__` before the block (which can return a value) and `__exit__` after, ensuring cleanup even if exceptions occur. Common uses include file handling, locks, database connections, and timers. They guarantee resource cleanup and make code more readable than try/finally blocks.

**Example Code:**


```
# Context manager demonstration
import time

class Timer:
    def __enter__(self):
        self.start = time.time()
        print("Timer started")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        elapsed = time.time() - self.start
        print(f"Timer stopped: {elapsed:.4f}s")
        return False  # Don't suppress exceptions

# Using the context manager
with Timer():
    total = sum(range(1000000))
    print(f"Sum: {total}")

# File handling is the most common use case
with open(__file__, 'r') as f:
    first_line = f.readline()
    print(f"\nFirst line of this file: {first_line[:50]}...")
```

### Question 16 – Exception Handling

**Question:**

Explain Python's exception handling mechanism. What are the roles of try/except/else/finally?

**Answer:**

`try` contains risky code. `except` catches and handles specific exceptions. `else` runs only if no exception occurred (good for code that should run on success). `finally` always runs for cleanup, whether an exception occurred or not. Exceptions are more Pythonic than return codes because they separate normal flow from error handling and can't be silently ignored.

**Example Code:**


```
# Exception handling demonstration
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
        return None
    except TypeError:
        print("Error: Invalid types for division")
        return None
    else:
        print("Division successful")
        return result
    finally:
        print("Cleanup: This always runs")

print("Test 1: divide(10, 2)")
print("Result:", divide(10, 2))

print("\nTest 2: divide(10, 0)")
print("Result:", divide(10, 0))

print("\nTest 3: divide('10', 2)")
print("Result:", divide('10', 2))
```

### Question 17 – Memory Management

**Question:**

How does Python manage memory? Explain reference counting and garbage collection.

**Answer:**

CPython uses **reference counting**: each object tracks how many references point to it. When the count hits zero, memory is freed immediately. **Cyclic garbage collection** detects and collects reference cycles (e.g., `a.next = b; b.prev = a`). The `del` statement removes a name binding; the object is freed when its refcount reaches zero. Note that `del` doesn't directly free memory - it just decrements the refcount.

**Example Code:**


```
# Memory management demonstration
import sys

class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

a = Node(1)
print("Refcount of a:", sys.getrefcount(a) - 1)  # -1 for getrefcount's own ref

b = a  # Another reference
print("Refcount after b=a:", sys.getrefcount(a) - 1)

c = a  # Yet another reference
print("Refcount after c=a:", sys.getrefcount(a) - 1)

del b
print("Refcount after del b:", sys.getrefcount(a) - 1)

# Circular reference (requires GC)
node1 = Node(1)
node2 = Node(2)
node1.next = node2
node2.next = node1  # Cycle!
print("\nCircular reference created - will be collected by GC")
```

### Question 18 – GIL and Concurrency

**Question:**

What is the GIL? How does it affect multi-threaded Python programs?

**Answer:**

The **Global Interpreter Lock (GIL)** ensures only one thread executes Python bytecode at a time in CPython. For **CPU-bound** tasks, threads don't provide speedup - use `multiprocessing` instead to bypass the GIL. For **I/O-bound** tasks, threads work well because the GIL is released during I/O waits. Alternatively, use `asyncio` for high-concurrency I/O-bound workloads.

**Example Code:**


```
# GIL demonstration
import threading
import time

def cpu_bound_task(n):
    """CPU-intensive task."""
    total = 0
    for i in range(n):
        total += i ** 2
    return total

# Sequential execution
start = time.time()
cpu_bound_task(5_000_000)
cpu_bound_task(5_000_000)
sequential_time = time.time() - start
print(f"Sequential: {sequential_time:.2f}s")

# Threaded execution (won't be faster due to GIL)
start = time.time()
t1 = threading.Thread(target=cpu_bound_task, args=(5_000_000,))
t2 = threading.Thread(target=cpu_bound_task, args=(5_000_000,))
t1.start()
t2.start()
t1.join()
t2.join()
threaded_time = time.time() - start
print(f"Threaded: {threaded_time:.2f}s (similar or slower due to GIL)")
```

### Question 19 – Deadlocks and Race Conditions

**Question:**

What are deadlocks and race conditions? How can you avoid them?

**Answer:**

A **deadlock** occurs when thread A holds lock1 and waits for lock2, while thread B holds lock2 and waits for lock1 - both are stuck. **Avoidance**: use consistent lock ordering, timeouts, or higher-level primitives like queues. A **race condition** happens when multiple threads access shared mutable state without proper synchronization, leading to unpredictable results. Use locks, thread-safe data structures, or avoid shared state.

**Example Code:**


```
# Deadlock and race condition demonstration
import threading

# Race condition example
counter = 0
lock = threading.Lock()

def increment_unsafe():
    """Race condition - not thread-safe."""
    global counter
    for _ in range(100000):
        counter += 1  # Not atomic!

def increment_safe():
    """Thread-safe with lock."""
    global counter
    for _ in range(100000):
        with lock:
            counter += 1

# Demonstrate race condition
counter = 0
t1 = threading.Thread(target=increment_unsafe)
t2 = threading.Thread(target=increment_unsafe)
t1.start()
t2.start()
t1.join()
t2.join()
print(f"Unsafe counter (expected 200000): {counter}")  # Often less than 200000!

# Demonstrate thread-safe version
counter = 0
t1 = threading.Thread(target=increment_safe)
t2 = threading.Thread(target=increment_safe)
t1.start()
t2.start()
t1.join()
t2.join()
print(f"Safe counter (expected 200000): {counter}")  # Always 200000
```

### Question 21 – Mutable Default Arguments

**Question:**

Why are mutable default arguments dangerous? How should you handle them?

**Answer:**

Default arguments are evaluated **once** at function definition time, not each call. Mutable defaults (list/dict) are reused across calls, so mutations leak between calls. **Fix**: Use `None` as the default and initialize inside the function, or use immutable defaults. This is one of Python's most common gotchas.

**Example Code:**


```
# Mutable default arguments demonstration
def buggy_append(item, lst=[]):
    """Bug: default list is shared across calls."""
    lst.append(item)
    return lst

print("Buggy version:")
print(buggy_append(1))  # [1]
print(buggy_append(2))  # [1, 2] - unexpected!
print(buggy_append(3))  # [1, 2, 3] - all calls share the same list!

def fixed_append(item, lst=None):
    """Fixed: create new list each time."""
    if lst is None:
        lst = []
    lst.append(item)
    return lst

print("\nFixed version:")
print(fixed_append(1))  # [1]
print(fixed_append(2))  # [2] - correct!
print(fixed_append(3))  # [3] - each call gets its own list
```

### Question 22 – Function Signatures

**Question:**

Explain positional-only, keyword-only, *args, and **kwargs parameters.

### Question 23 – Closures and Late Binding

**Question:**

What is the late binding closure problem? How do you fix it?

**Answer:**

Loop variables are looked up when the lambda/function is **called**, not when it's defined. All closures may see the final loop value. **Fix 1**: Capture the value via a default argument (e.g., `lambda x=i: x`). **Fix 2**: Use `functools.partial`. This is a common source of bugs in loops creating functions.

**Example Code:**


```
# Closures and late binding demonstration
# Bug: all lambdas see the final value of i
funcs = []
for i in range(3):
    funcs.append(lambda: i)

print("Buggy version:")
print([f() for f in funcs])  # [2, 2, 2] - all see i=2!

# Fix 1: default argument captures the value
funcs_fixed = []
for i in range(3):
    funcs_fixed.append(lambda x=i: x)

print("\nFixed with default argument:")
print([f() for f in funcs_fixed])  # [0, 1, 2] - correct!

# Fix 2: functools.partial
from functools import partial

funcs_partial = []
for i in range(3):
    funcs_partial.append(partial(lambda x: x, i))

print("\nFixed with partial:")
print([f() for f in funcs_partial])  # [0, 1, 2]
```

### Question 24 – Decorators

**Question:**

What are decorators? How do you write one? Why use `functools.wraps`?

**Answer:**

A decorator is a callable that takes a function and returns a new callable. Used for cross-cutting concerns like logging, timing, authentication, and caching. Use `functools.wraps` to preserve the wrapped function's metadata (`__name__`, `__doc__`, etc.), which is important for debugging and introspection.

**Example Code:**


```
# Decorators demonstration
import functools
import time

def timing_decorator(func):
    """Decorator that times function execution."""
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        elapsed = time.time() - start
        print(f"{func.__name__} took {elapsed:.4f}s")
        return result
    return wrapper

@timing_decorator
def slow_computation(n):
    """Compute sum of squares."""
    return sum(i**2 for i in range(n))

result = slow_computation(100000)
print("Result:", result)
print("Function name:", slow_computation.__name__)  # Preserved by @wraps
print("Function doc:", slow_computation.__doc__)    # Preserved by @wraps
```

### Question 27 – Contextlib

**Question:**

How does `@contextlib.contextmanager` work? When should you use it?

**Answer:**

`@contextmanager` lets you write a generator-based context manager. Yield once: code before `yield` is `__enter__`, code after is `__exit__`. More concise than writing a full class for simple setup/teardown patterns. The `try/finally` around `yield` ensures cleanup even on exceptions.

**Example Code:**


```
# Contextlib demonstration
from contextlib import contextmanager

@contextmanager
def temporary_value(var_dict, key, temp_value):
    """Temporarily set a dict value, then restore."""
    old_value = var_dict.get(key)
    var_dict[key] = temp_value
    try:
        yield  # Code in 'with' block runs here
    finally:
        # Cleanup: restore old value
        if old_value is None:
            var_dict.pop(key, None)
        else:
            var_dict[key] = old_value

config = {"debug": False, "verbose": True}
print("Before:", config)

with temporary_value(config, "debug", True):
    print("Inside context:", config)  # debug=True

print("After:", config)  # debug=False (restored)
```

### Question 29 – Descriptors

**Question:**

**Answer:**

### Question 31 – Dynamic Attribute Access

**Question:**

What's the difference between `__getattr__` and `__getattribute__`? When should you use each?

**Answer:**

`__getattr__` is called only when an attribute is **not found** through normal lookup. `__getattribute__` intercepts **every** attribute access. Use `__getattr__` for proxies, lazy loading, or default values. Avoid `__getattribute__` unless necessary - it's easy to cause infinite recursion. Always use `super().__getattribute__()` to delegate to avoid recursion.

**Example Code:**


```
# Dynamic attribute access demonstration
class LazyLoader:
    def __init__(self):
        self.data = {"x": 10, "y": 20}

    def __getattr__(self, name):
        """Called when attribute not found normally."""
        print(f"__getattr__ called for '{name}'")
        if name in self.data:
            return self.data[name]
        raise AttributeError(f"No attribute '{name}'")

obj = LazyLoader()
print("Accessing x:", obj.x)  # Calls __getattr__
print("Accessing data:", obj.data)  # Normal lookup, doesn't call __getattr__

# __getattribute__ example
class LoggingProxy:
    def __init__(self, target):
        # Use object.__setattr__ to avoid infinite recursion
        object.__setattr__(self, "_target", target)

    def __getattribute__(self, name):
        """Intercepts all attribute access."""
        if name == "_target":
            return object.__getattribute__(self, name)
        print(f"Accessing attribute: {name}")
        target = object.__getattribute__(self, "_target")
        return getattr(target, name)

class Data:
    value = 42

proxy = LoggingProxy(Data())
print("\nProxy access:")
print(proxy.value)
```

### Question 34 – Rich Comparisons

**Question:**

How do you implement comparison operators for custom classes? What is `@total_ordering`?

**Answer:**

Implement `__eq__` for equality and one ordering method (e.g., `__lt__`). The `@total_ordering` decorator from `functools` auto-generates the remaining comparison methods (`__le__`, `__gt__`, `__ge__`). You must ensure the comparison defines a consistent total order for sorting to work correctly.

**Example Code:**


```
# Rich comparisons demonstration
from functools import total_ordering

@total_ordering
class Version:
    def __init__(self, major, minor):
        self.major = major
        self.minor = minor

    def __eq__(self, other):
        if not isinstance(other, Version):
            return NotImplemented
        return (self.major, self.minor) == (other.major, other.minor)

    def __lt__(self, other):
        if not isinstance(other, Version):
            return NotImplemented
        return (self.major, self.minor) < (other.major, other.minor)

    def __repr__(self):
        return f"Version({self.major}.{self.minor})"

v1 = Version(1, 2)
v2 = Version(1, 3)
v3 = Version(2, 0)

print("v1 < v2:", v1 < v2)  # True
print("v2 <= v3:", v2 <= v3)  # True (auto-generated)
print("v3 > v1:", v3 > v1)  # True (auto-generated)
print("Sorted:", sorted([v3, v1, v2]))
```

### Question 35 – Hash and Equality for Custom Types

**Question:**

What are the requirements for implementing `__hash__` and `__eq__` for dict/set keys?

**Answer:**

For dict/set keys, implement both `__hash__` and `__eq__`. **Critical requirement**: If `a == b`, then `hash(a) == hash(b)`. Only hash **immutable** fields - mutating hashed objects breaks dicts/sets. Hashable objects should be immutable or at least have immutable hash-relevant fields.

**Example Code:**


```
# Hash and equality demonstration
class ImmutablePoint:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    @property
    def x(self):
        return self._x

    @property
    def y(self):
        return self._y

    def __eq__(self, other):
        if not isinstance(other, ImmutablePoint):
            return False
        return (self.x, self.y) == (other.x, other.y)

    def __hash__(self):
        return hash((self.x, self.y))

    def __repr__(self):
        return f"ImmutablePoint({self.x}, {self.y})"

p1 = ImmutablePoint(1, 2)
p2 = ImmutablePoint(1, 2)
p3 = ImmutablePoint(3, 4)

print("p1 == p2:", p1 == p2)  # True
print("hash(p1) == hash(p2):", hash(p1) == hash(p2))  # True

# Can use as dict key
points_map = {p1: "origin area", p3: "other"}
print("Dict lookup with p2:", points_map[p2])  # "origin area" (p2 == p1)

# Can use in set
point_set = {p1, p2, p3}
print("Set (p1 and p2 are same):", point_set)


```

### Question 36 – Iterator Exhaustion

**Answer:**

Iterators are **one-shot**: once consumed, they're exhausted and can't be reused. ### Question 37 – Generator Cleanup

**Question:**

How do you ensure cleanup in generators?

**Answer:**

Use `try/finally` in generators to ensure cleanup even if iteration stops early. The runtime calls `.close()` (which triggers `finally`) when the generator is garbage-collected or explicitly closed. This is essential for managing resources like files, locks, or database connections in generators.

**Example Code:**


```
# Generator cleanup demonstration
def managed_resource():
    """Generator that ensures cleanup."""
    print("Acquiring resource")
    try:
        yield "resource"
    finally:
        print("Releasing resource (cleanup)")

# Normal completion
print("Normal completion:")
for res in managed_resource():
    print(f"Using: {res}")

# Early termination
print("\nEarly termination:")
gen = managed_resource()
print(next(gen))
gen.close()  # Explicitly close, triggers finally

# Automatic cleanup on garbage collection
print("\nAutomatic cleanup:")
gen = managed_resource()
next(gen)
del gen  # Triggers cleanup when garbage collected
```

### Question 38 – `yield from`

**Question:**

What does `yield from` do? When should you use it?

**Answer:**

`yield from subgen` delegates to a sub-generator, forwarding all values, `.send()` calls, and exceptions bidirectionally. It's cleaner and more efficient than manually looping and yielding. Use it for composing generators, flattening nested structures, or implementing coroutines.

**Example Code:**


```
# yield from demonstration
def sub_generator():
    yield 1
    yield 2
    yield 3

def main_generator():
    yield "start"
    yield from sub_generator()  # Delegates to sub_generator
    yield "end"

print("Using yield from:")
for value in main_generator():
    print(value)

# Compare with manual approach
def main_generator_manual():
    yield "start"
    for value in sub_generator():  # Manual delegation
        yield value
    yield "end"

print("\nManual approach (same result):")
for value in main_generator_manual():
    print(value)

# Flattening nested lists
def flatten(nested):
    for item in nested:
        if isinstance(item, list):
            yield from flatten(item)  # Recursive delegation
        else:
            yield item

nested_list = [1, [2, 3, [4, 5]], 6]
print("\nFlattening:", list(flatten(nested_list)))
```

### Question 39 – Streaming Pipelines

**Question:**

How do you build memory-efficient streaming pipelines?

**Answer:**

Process large files/streams line-by-line or chunk-by-chunk using generators and `itertools`. This keeps memory usage bounded regardless of input size. Combine with context managers for safe resource handling. Each stage is a generator that yields processed items to the next stage.

**Example Code:**


```
# Streaming pipeline demonstration
def read_lines(filename):
    """Generator that yields lines from a file."""
    with open(filename) as f:
        for line in f:
            yield line.strip()

def filter_lines(lines, keyword):
    """Generator that filters lines containing keyword."""
    for line in lines:
        if keyword in line:
            yield line

def transform_lines(lines):
    """Generator that transforms lines."""
    for line in lines:
        yield line.upper()

def process_large_file(filename, keyword):
    """Streaming pipeline: read -> filter -> transform -> process."""
    lines = read_lines(filename)
    filtered = filter_lines(lines, keyword)
    transformed = transform_lines(filtered)

    for line in transformed:
        print("Processed:", line)

# Example with in-memory data (simulating file)
def simulate_file_lines():
    """Simulate reading lines from a file."""
    data = ["error: failed", "info: success", "error: timeout", "debug: trace"]
    for line in data:
        yield line

print("Streaming pipeline example:")
filtered = filter_lines(simulate_file_lines(), "error")
transformed = transform_lines(filtered)
for line in transformed:
    print("Result:", line)
```

### Question 40 – Batching

**Question:**

How do you batch items from an iterator?

**Answer:**

Yield items in fixed-size batches for downstream consumers. Useful for database bulk inserts, network calls with batch APIs, or parallel processing. Remember to yield any remaining items in the final partial batch.

**Example Code:**


```
# Batching demonstration
def batch_iterator(iterable, batch_size):
    """Yield items in batches of batch_size."""
    batch = []
    for item in iterable:
        batch.append(item)
        if len(batch) == batch_size:
            yield batch
            batch = []
    if batch:  # Yield remaining items
        yield batch

print("Batching example:")
data = range(10)
for i, batch in enumerate(batch_iterator(data, 3)):
    print(f"Batch {i + 1}:", batch)

# Using itertools for more efficient batching
from itertools import islice

def batch_iterator_itertools(iterable, batch_size):
    """More efficient batching using itertools."""
    iterator = iter(iterable)
    while True:
        batch = list(islice(iterator, batch_size))
        if not batch:
            break
        yield batch

print("\nUsing itertools:")
for i, batch in enumerate(batch_iterator_itertools(range(10), 3)):
    print(f"Batch {i + 1}:", batch)
```

Compare Python's concurrency models: threads, processes, and async IO.

### Question 42 – `concurrent.futures`

**Example Code:**


```
# concurrent.futures demonstration
from concurrent.futures import ThreadPoolExecutor, as_completed

def task(n):
    """Task that may fail."""
    if n == 3:
        raise ValueError(f"Task {n} failed")
    return n * n

print("Using concurrent.futures:")
with ThreadPoolExecutor(max_workers=4) as executor:
    # Submit tasks and get futures
    futures = {executor.submit(task, i): i for i in range(5)}

    # Process results as they complete
    for future in as_completed(futures):
        i = futures[future]
        try:
            result = future.result()
            print(f"Task {i} result: {result}")
        except Exception as e:
            print(f"Task {i} failed: {e}")
```

### Question 44 – Shared State and Locks

How do you safely share state between threads?

**Answer:**

The GIL doesn't protect your data structures from race conditions. Use `threading.Lock`, `RLock`, or higher-level primitives like queues. Always acquire locks in consistent order to avoid deadlocks. For simple cases, use `queue.Queue` which is thread-safe.

**Example Code:**


```
# Shared state and locks demonstration
import threading

counter = 0
lock = threading.Lock()

def increment_unsafe():
    """Unsafe increment - race condition."""
    global counter
    for _ in range(100000):
        counter += 1

def increment_safe():
    """Safe increment with lock."""
    global counter
    for _ in range(100000):
        with lock:
            counter += 1

# Safe version with lock
print("Using locks:")
counter = 0
threads = [threading.Thread(target=increment_safe) for _ in range(5)]
for t in threads:
    t.start()
for t in threads:
    t.join()

print(f"Final counter (with lock): {counter}")  # Should be 500000
```

### Question 45 – Rate Limiting / Throttling

**Question:**

How do you implement rate limiting?

**Answer:**

Track timestamps or use a token bucket algorithm. Use locks or async primitives (`asyncio.Semaphore`) to enforce limits. For async code, `asyncio.Semaphore` limits concurrent operations. For rate limiting over time, implement token bucket or sliding window algorithms.

**Example Code:**


```
# Rate limiting demonstration
import asyncio
import time

class RateLimiter:
    """Token bucket rate limiter."""
    def __init__(self, rate, per):
        """Allow 'rate' calls per 'per' seconds."""
        self.rate = rate
        self.per = per
        self.allowance = rate
        self.last_check = time.time()
        self.lock = asyncio.Lock()

    async def acquire(self):
        """Acquire permission to proceed."""
        async with self.lock:
            current = time.time()
            elapsed = current - self.last_check
            self.last_check = current
            self.allowance += elapsed * (self.rate / self.per)
            if self.allowance > self.rate:
                self.allowance = self.rate

            if self.allowance < 1:
                sleep_time = (1 - self.allowance) * (self.per / self.rate)
                await asyncio.sleep(sleep_time)
                self.allowance = 0
            else:
                self.allowance -= 1

async def make_request(limiter, i):
    await limiter.acquire()
    print(f"Request {i} at {time.time():.2f}")

async def test_rate_limiter():
    limiter = RateLimiter(rate=2, per=1)  # 2 requests per second
    tasks = [make_request(limiter, i) for i in range(5)]
    await asyncio.gather(*tasks)

print("Rate limiting example:")
asyncio.run(test_rate_limiter())


```

### Question 48 – Type Hints

**Question:**

How do type hints improve Python code? When should you use them?

**Answer:**

Type hints improve readability, IDE support (autocomplete, refactoring), and enable early error detection with static type checkers like `mypy`. They're documentation that's checked by tools. Introduce gradually around public APIs and critical modules. Use `typing` module for complex types (List, Dict, Optional, Union, etc.).

**Example Code:**


```
# Type hints demonstration
from typing import List, Optional, Dict, Union

def process_items(items: List[int], threshold: int) -> Dict[str, int]:
    """Process items and return summary with type hints."""
    above = sum(1 for x in items if x > threshold)
    below = len(items) - above
    return {"above": above, "below": below}

def find_user(user_id: int) -> Optional[Dict[str, str]]:
    """Find user by ID, return None if not found."""
    users = {1: {"name": "Alice"}, 2: {"name": "Bob"}}
    return users.get(user_id)

def parse_value(value: Union[int, str]) -> int:
    """Parse value that can be int or str."""
    if isinstance(value, str):
        return int(value)
    return value

# Usage
result = process_items([1, 5, 10, 15], 7)
print("Summary:", result)

user = find_user(1)
print("User:", user)

print("Parse '42':", parse_value("42"))
print("Parse 42:", parse_value(42))
```

### Question 49 – Performance Analysis

**Question:**

How do you profile and optimize Python code?

**Answer:**

Use `cProfile` for function-level profiling to find hotspots. Use line profilers (`line_profiler`) for line-by-line analysis. Use `timeit` for microbenchmarks. **Strategy**: Address algorithmic complexity first (O(n²) → O(n)), then micro-optimize hot paths. Profile before optimizing - don't guess!

**Example Code:**


```
# Performance analysis demonstration
import cProfile
import pstats
from io import StringIO
import timeit

def slow_function():
    """Inefficient implementation."""
    total = 0
    for i in range(100000):
        total += i ** 2
    return total

def fast_function():
    """More efficient implementation."""
    return sum(i ** 2 for i in range(100000))

# Profile slow_function
profiler = cProfile.Profile()
profiler.enable()
slow_function()
profiler.disable()

s = StringIO()
ps = pstats.Stats(profiler, stream=s).sort_stats("cumulative")
ps.print_stats(5)
print("Profile output:")
print(s.getvalue()[:200], "...")

# Benchmark comparison
slow_time = timeit.timeit(slow_function, number=10)
fast_time = timeit.timeit(fast_function, number=10)

print(f"\nBenchmark:")
print(f"Slow: {slow_time:.4f}s")
print(f"Fast: {fast_time:.4f}s")
print(f"Speedup: {slow_time / fast_time:.2f}x")
```

### Question 50 – Beyond Pure Python

**Question:**

What options exist for improving Python performance beyond pure Python?

**Answer:**

**Optimized libraries**: NumPy, pandas for numerical/data work (C/Fortran under the hood). **C extensions**: Write performance-critical code in C. **Cython**: Python-like syntax compiled to C. **Numba**: JIT compiler for numerical code. **PyPy**: Alternative Python interpreter with JIT. **Separate service**: Move heavy computation to faster languages (Go, Rust, C++).

**Example Code:**


```
# Beyond pure Python demonstration
import numpy as np
import timeit

# Pure Python
def sum_squares_python(n):
    """Pure Python implementation."""
    return sum(i ** 2 for i in range(n))

# NumPy (much faster for large n)
def sum_squares_numpy(n):
    """NumPy implementation - uses optimized C code."""
    return np.sum(np.arange(n) ** 2)

# Benchmark
n = 100000
python_time = timeit.timeit(lambda: sum_squares_python(n), number=10)
numpy_time = timeit.timeit(lambda: sum_squares_numpy(n), number=10)

print("Performance comparison:")
print(f"Pure Python: {python_time:.4f}s")
print(f"NumPy: {numpy_time:.4f}s")
print(f"Speedup: {python_time / numpy_time:.1f}x")

# NumPy array operations
arr = np.arange(1000000)
print(f"\nNumPy array sum: {np.sum(arr ** 2)}")
print("NumPy provides massive speedups for numerical operations!")
```
